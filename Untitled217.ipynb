```python
import os
import time
import json
import yaml
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense

# ==============================================================
# 1. CONFIGURATION
# ==============================================================

START_TIME = time.time()

SEED = 42

np.random.seed(SEED)
tf.random.set_seed(SEED)

rng = np.random.default_rng(SEED)

OUTPUT_DIR = r"C:\Users\sagni\Downloads\AI-Based Municipal Waste Management and Conservancy Workforce"

ORIGINAL_DATASET = os.path.join(
    OUTPUT_DIR,
    "number_of conservancy_workers_engaged_in_conservancy_garbage_and_debris_disposal_and_other_particulars_2017_18.csv"
)

SYNTHETIC_DATASET = os.path.join(
    OUTPUT_DIR,
    "synthetic_municipal_dataset.csv"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

sns.set_theme(style="whitegrid")

plt.rcParams["figure.dpi"] = 120
plt.rcParams["savefig.dpi"] = 300
plt.rcParams["font.size"] = 11

# PSO Hyperparameters

SWARM_SIZE = 40
ITERATIONS = 150

W_MAX = 0.9
W_MIN = 0.4

C1 = 1.5
C2 = 1.5

POSITION_LIMIT = 5.0
VELOCITY_LIMIT = 0.3

REGULARIZATION = 0.0001

HIDDEN_LAYER_1 = 16
HIDDEN_LAYER_2 = 8

print("=" * 75)
print("PSO-BASED MUNICIPAL WASTE MANAGEMENT")
print("PARTICLE SWARM OPTIMIZATION")
print("=" * 75)

# ==============================================================
# 2. LOAD DATASET
# ==============================================================

if not os.path.exists(SYNTHETIC_DATASET):

    raise FileNotFoundError(
        "Synthetic dataset not found. "
        "Run the baseline model training code first."
    )

df = pd.read_csv(SYNTHETIC_DATASET)

print("\nDataset Shape:", df.shape)

print("\nDataset Preview:")
print(df.head())

df.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "pso_dataset.csv"
    ),
    index=False
)

# ==============================================================
# 3. FEATURE AND TARGET SELECTION
# ==============================================================

FEATURES = [

    "disposal_sites",

    "other_workers",

    "daily_waste_tonnes",

    "population_density",

    "working_hours",

    "worker_productivity",

    "equipment_efficiency",

    "absenteeism_rate",

    "transport_distance_km",

    "seasonal_demand_factor"
]

TARGET = "required_workers"

required_columns = FEATURES + [

    TARGET,

    "available_workers",

    "daily_wage"
]

missing_columns = [

    col for col in required_columns

    if col not in df.columns
]

if missing_columns:

    raise ValueError(
        f"Missing columns: {missing_columns}"
    )

if not np.isfinite(
    df[required_columns].to_numpy(dtype=float)
).all():

    raise ValueError(
        "Missing or non-finite dataset values detected."
    )

X = df[FEATURES]
y = df[TARGET]

print("\nInput Features:", FEATURES)
print("\nTarget:", TARGET)

# ==============================================================
# 4. TRAIN, VALIDATION AND TEST SPLIT
# ==============================================================

# Same split as baseline and AIS.

X_train_full, X_test, y_train_full, y_test = (
    train_test_split(

        X,

        y,

        test_size=0.20,

        random_state=SEED
    )
)

X_train, X_val, y_train, y_val = (
    train_test_split(

        X_train_full,

        y_train_full,

        test_size=0.20,

        random_state=SEED
    )
)

print("\nTraining Samples:", len(X_train))
print("Validation Samples:", len(X_val))
print("Testing Samples:", len(X_test))

# ==============================================================
# 5. FEATURE AND TARGET SCALING
# ==============================================================

feature_scaler = StandardScaler()

X_train_scaled = feature_scaler.fit_transform(
    X_train
).astype(np.float32)

X_val_scaled = feature_scaler.transform(
    X_val
).astype(np.float32)

X_test_scaled = feature_scaler.transform(
    X_test
).astype(np.float32)

target_scaler = StandardScaler()

y_train_scaled = target_scaler.fit_transform(

    y_train.to_numpy().reshape(-1, 1)

).flatten().astype(np.float32)

y_val_scaled = target_scaler.transform(

    y_val.to_numpy().reshape(-1, 1)

).flatten().astype(np.float32)

joblib.dump(

    feature_scaler,

    os.path.join(
        OUTPUT_DIR,
        "pso_scaler.pkl"
    )
)

joblib.dump(

    target_scaler,

    os.path.join(
        OUTPUT_DIR,
        "pso_target_scaler.pkl"
    )
)

# ==============================================================
# 6. NEURAL NETWORK ARCHITECTURE
# ==============================================================

INPUT_SIZE = len(FEATURES)

WEIGHT_SHAPES = [

    (INPUT_SIZE, HIDDEN_LAYER_1),

    (HIDDEN_LAYER_1,),

    (HIDDEN_LAYER_1, HIDDEN_LAYER_2),

    (HIDDEN_LAYER_2,),

    (HIDDEN_LAYER_2, 1),

    (1,)
]

TOTAL_PARAMETERS = sum(

    int(np.prod(shape))

    for shape in WEIGHT_SHAPES
)

print("\nNeural Network Architecture:")

print(
    INPUT_SIZE,
    "->",
    HIDDEN_LAYER_1,
    "->",
    HIDDEN_LAYER_2,
    "-> 1"
)

print("Total PSO Parameters:", TOTAL_PARAMETERS)

# ==============================================================
# 7. PARTICLE REPRESENTATION
# ==============================================================

# Every particle represents all weights
# and biases of the neural network.

def decode_particle(particle):

    weights = []

    start = 0

    for shape in WEIGHT_SHAPES:

        size = int(np.prod(shape))

        weights.append(

            particle[
                start:start + size
            ].reshape(shape)
        )

        start += size

    return weights

# ==============================================================
# 8. NEURAL NETWORK FORWARD PROPAGATION
# ==============================================================

def relu(x):

    return np.maximum(x, 0)


def forward_pass(X_data, particle):

    W1, b1, W2, b2, W3, b3 = (
        decode_particle(particle)
    )

    hidden1 = relu(
        X_data @ W1 + b1
    )

    hidden2 = relu(
        hidden1 @ W2 + b2
    )

    output = (
        hidden2 @ W3 + b3
    )

    return output.flatten()

# ==============================================================
# 9. PSO FITNESS FUNCTION
# ==============================================================

def fitness_function(

    particle,

    X_data,

    y_data,

    regularize=True
):

    predictions = forward_pass(

        X_data,

        particle
    )

    mse = np.mean(

        (y_data - predictions) ** 2
    )

    penalty = (

        REGULARIZATION
        * np.mean(particle ** 2)

        if regularize else 0.0
    )

    return float(mse + penalty)

# ==============================================================
# 10. INITIALIZE PARTICLE SWARM
# ==============================================================

def create_particle():

    parameters = []

    for shape in WEIGHT_SHAPES:

        if len(shape) == 2:

            std = np.sqrt(
                2.0 / shape[0]
            )

            values = rng.normal(

                0,

                std,

                size=shape
            )

        else:

            values = np.zeros(shape)

        parameters.append(
            values.flatten()
        )

    return np.concatenate(

        parameters

    ).astype(np.float32)


positions = np.array([

    create_particle()

    for _ in range(SWARM_SIZE)

], dtype=np.float32)

velocities = rng.uniform(

    -0.1,

    0.1,

    size=(
        SWARM_SIZE,
        TOTAL_PARAMETERS
    )

).astype(np.float32)

personal_best_positions = positions.copy()

personal_best_scores = np.array([

    fitness_function(

        particle,

        X_train_scaled,

        y_train_scaled
    )

    for particle in positions
])

best_index = np.argmin(
    personal_best_scores
)

global_best_position = (
    personal_best_positions[
        best_index
    ].copy()
)

global_best_score = float(

    personal_best_scores[
        best_index
    ]
)

best_validation_position = (
    global_best_position.copy()
)

best_validation_loss = fitness_function(

    global_best_position,

    X_val_scaled,

    y_val_scaled,

    regularize=False
)

best_iteration = 0

convergence_history = []

fitness_evaluations = SWARM_SIZE

print("\nSwarm Shape:", positions.shape)

print(
    "Initial Global Best Fitness:",
    global_best_score
)

# ==============================================================
# 11. PARTICLE SWARM OPTIMIZATION
# ==============================================================

print("\nStarting PSO Optimization...")
print("=" * 75)

optimization_start = time.time()

for iteration in range(

    1,

    ITERATIONS + 1
):

    # ----------------------------------------------------------
    # A. Adaptive inertia weight
    # ----------------------------------------------------------

    inertia_weight = (

        W_MAX

        - (W_MAX - W_MIN)

        * (iteration - 1)

        / max(ITERATIONS - 1, 1)
    )

    # ----------------------------------------------------------
    # B. Generate random coefficients
    # ----------------------------------------------------------

    r1 = rng.random(

        size=positions.shape

    ).astype(np.float32)

    r2 = rng.random(

        size=positions.shape

    ).astype(np.float32)

    # ----------------------------------------------------------
    # C. Cognitive component
    # ----------------------------------------------------------

    cognitive_component = (

        C1

        * r1

        * (
            personal_best_positions
            - positions
        )
    )

    # ----------------------------------------------------------
    # D. Social component
    # ----------------------------------------------------------

    social_component = (

        C2

        * r2

        * (
            global_best_position
            - positions
        )
    )

    # ----------------------------------------------------------
    # E. Update particle velocities
    # ----------------------------------------------------------

    velocities = (

        inertia_weight * velocities

        + cognitive_component

        + social_component
    )

    velocities = np.clip(

        velocities,

        -VELOCITY_LIMIT,

        VELOCITY_LIMIT
    )

    # ----------------------------------------------------------
    # F. Update particle positions
    # ----------------------------------------------------------

    positions = (

        positions + velocities
    )

    positions = np.clip(

        positions,

        -POSITION_LIMIT,

        POSITION_LIMIT
    )

    # ----------------------------------------------------------
    # G. Evaluate particle fitness
    # ----------------------------------------------------------

    fitness_scores = np.array([

        fitness_function(

            particle,

            X_train_scaled,

            y_train_scaled
        )

        for particle in positions
    ])

    fitness_evaluations += SWARM_SIZE

    # ----------------------------------------------------------
    # H. Update personal best
    # ----------------------------------------------------------

    improved = (

        fitness_scores

        < personal_best_scores
    )

    personal_best_positions[improved] = (

        positions[improved]
    )

    personal_best_scores[improved] = (

        fitness_scores[improved]
    )

    # ----------------------------------------------------------
    # I. Update global best
    # ----------------------------------------------------------

    current_best_index = np.argmin(

        personal_best_scores
    )

    current_best_score = float(

        personal_best_scores[
            current_best_index
        ]
    )

    if current_best_score < global_best_score:

        global_best_score = (
            current_best_score
        )

        global_best_position = (

            personal_best_positions[
                current_best_index
            ].copy()
        )

    # ----------------------------------------------------------
    # J. Validation-based checkpoint selection
    # ----------------------------------------------------------

    training_loss = fitness_function(

        global_best_position,

        X_train_scaled,

        y_train_scaled,

        regularize=False
    )

    validation_loss = fitness_function(

        global_best_position,

        X_val_scaled,

        y_val_scaled,

        regularize=False
    )

    if validation_loss < best_validation_loss:

        best_validation_loss = (
            validation_loss
        )

        best_validation_position = (
            global_best_position.copy()
        )

        best_iteration = iteration

    # ----------------------------------------------------------
    # K. Record convergence history
    # ----------------------------------------------------------

    convergence_history.append({

        "Iteration": iteration,

        "Inertia_Weight": float(
            inertia_weight
        ),

        "Global_Best_Fitness": float(
            global_best_score
        ),

        "Average_Swarm_Fitness": float(
            np.mean(fitness_scores)
        ),

        "Training_Loss": float(
            training_loss
        ),

        "Validation_Loss": float(
            validation_loss
        ),

        "Best_Validation_Loss": float(
            best_validation_loss
        ),

        "Swarm_Diversity": float(

            np.mean(
                np.std(
                    positions,
                    axis=0
                )
            )
        ),

        "Average_Velocity": float(

            np.mean(
                np.abs(velocities)
            )
        ),

        "Improved_Particles": int(
            np.sum(improved)
        ),

        "Fitness_Evaluations": int(
            fitness_evaluations
        )
    })

    if iteration == 1 or iteration % 10 == 0:

        print(

            f"Iteration {iteration:03d}/{ITERATIONS} | "

            f"Inertia: {inertia_weight:.4f} | "

            f"Fitness: {global_best_score:.6f} | "

            f"Validation: {validation_loss:.6f} | "

            f"Best Validation: {best_validation_loss:.6f}"
        )

optimization_time = (
    time.time() - optimization_start
)

best_particle = (
    best_validation_position.copy()
)

print("\nPSO Optimization Completed.")

print("Best Iteration:", best_iteration)

print(
    "Best Validation Loss:",
    best_validation_loss
)

print(
    "Optimization Time:",
    round(optimization_time, 2),
    "seconds"
)

# ==============================================================
# 12. SAVE PSO CONVERGENCE HISTORY
# ==============================================================

history_df = pd.DataFrame(
    convergence_history
)

history_df.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "pso_convergence_history.csv"
    ),

    index=False
)

np.save(

    os.path.join(
        OUTPUT_DIR,
        "pso_best_particle.npy"
    ),

    best_particle
)

np.save(

    os.path.join(
        OUTPUT_DIR,
        "pso_global_best_particle.npy"
    ),

    global_best_position
)

np.savez_compressed(

    os.path.join(
        OUTPUT_DIR,
        "pso_swarm_state.npz"
    ),

    positions=positions,

    velocities=velocities,

    personal_best_positions=personal_best_positions,

    personal_best_scores=personal_best_scores,

    global_best_position=global_best_position,

    best_validation_position=best_particle
)

# ==============================================================
# 13. CREATE KERAS MODEL USING PSO WEIGHTS
# ==============================================================

pso_model = Sequential([

    Input(
        shape=(INPUT_SIZE,)
    ),

    Dense(
        HIDDEN_LAYER_1,
        activation="relu"
    ),

    Dense(
        HIDDEN_LAYER_2,
        activation="relu"
    ),

    Dense(
        1,
        activation="linear"
    )
])

# All weights are obtained from PSO.
# No backpropagation or model.fit() is used.

pso_model.set_weights(
    decode_particle(best_particle)
)

pso_model.compile(

    optimizer="adam",

    loss="mse",

    metrics=["mae"]
)

pso_model.save(

    os.path.join(
        OUTPUT_DIR,
        "pso_model.h5"
    )
)

print("\nPSO H5 Model Saved.")

# ==============================================================
# 14. SAVE PKL MODEL BUNDLE
# ==============================================================

pso_model_bundle = {

    "algorithm": "Particle Swarm Optimization",

    "best_particle": best_particle,

    "global_best_particle": global_best_position,

    "weight_shapes": WEIGHT_SHAPES,

    "weights": decode_particle(
        best_particle
    ),

    "feature_scaler": feature_scaler,

    "target_scaler": target_scaler,

    "features": FEATURES,

    "target": TARGET,

    "architecture": [

        INPUT_SIZE,

        HIDDEN_LAYER_1,

        HIDDEN_LAYER_2,

        1
    ],

    "best_iteration": best_iteration,

    "data_type": "synthetic_simulation"
}

joblib.dump(

    pso_model_bundle,

    os.path.join(
        OUTPUT_DIR,
        "pso_model.pkl"
    )
)

print("PSO PKL Model Saved.")

# ==============================================================
# 15. WORKFORCE PREDICTION FUNCTION
# ==============================================================

def predict_workforce(

    X_scaled,

    X_original
):

    scaled_predictions = forward_pass(

        X_scaled,

        best_particle
    )

    predictions = target_scaler.inverse_transform(

        scaled_predictions.reshape(-1, 1)

    ).flatten()

    minimum_workers = (

        2 * X_original[
            "disposal_sites"
        ].to_numpy()
    )

    predictions = np.maximum(

        np.ceil(predictions),

        minimum_workers
    )

    return predictions.astype(int)


train_predictions = predict_workforce(

    X_train_scaled,

    X_train
)

val_predictions = predict_workforce(

    X_val_scaled,

    X_val
)

test_predictions = predict_workforce(

    X_test_scaled,

    X_test
)

# ==============================================================
# 16. PERFORMANCE EVALUATION
# ==============================================================

def evaluate_model(actual, predicted):

    actual = np.asarray(
        actual
    ).flatten()

    predicted = np.asarray(
        predicted
    ).flatten()

    mae = mean_absolute_error(

        actual,

        predicted
    )

    mse = mean_squared_error(

        actual,

        predicted
    )

    rmse = np.sqrt(mse)

    r2 = r2_score(

        actual,

        predicted
    )

    mape = np.mean(

        np.abs(

            (actual - predicted)

            / np.maximum(
                np.abs(actual),
                1
            )
        )

    ) * 100

    return {

        "MAE": float(mae),

        "MSE": float(mse),

        "RMSE": float(rmse),

        "R2_Score": float(r2),

        "R2_Percentage": float(
            r2 * 100
        ),

        "MAPE_Percentage": float(mape)
    }


train_metrics = evaluate_model(

    y_train,

    train_predictions
)

validation_metrics = evaluate_model(

    y_val,

    val_predictions
)

test_metrics = evaluate_model(

    y_test,

    test_predictions
)

print("\n" + "=" * 75)
print("PSO TEST PERFORMANCE")
print("=" * 75)

for metric, value in test_metrics.items():

    print(

        f"{metric:<25}: {value:.6f}"
    )

# ==============================================================
# 17. SAVE RESULT CSV
# ==============================================================

result_df = pd.DataFrame([

    {

        "Model": "PSO-ANN",

        "Dataset": "Training",

        **train_metrics
    },

    {

        "Model": "PSO-ANN",

        "Dataset": "Validation",

        **validation_metrics
    },

    {

        "Model": "PSO-ANN",

        "Dataset": "Testing",

        **test_metrics
    }
])

result_df["Evaluation_Type"] = (
    "Synthetic Simulation"
)

result_df.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "pso_result.csv"
    ),

    index=False
)

print("\nPSO Result DataFrame:")

print(
    result_df.round(4).to_string(
        index=False
    )
)

# ==============================================================
# 18. CREATE PREDICTION DATAFRAME
# ==============================================================

prediction_df = df.loc[
    X_test.index
].copy()

prediction_df.insert(

    0,

    "Scenario_ID",

    X_test.index.to_numpy() + 1
)

prediction_df["Actual_Required_Workers"] = (
    y_test.to_numpy()
)

prediction_df["PSO_Predicted_Required_Workers"] = (
    test_predictions
)

prediction_df["Absolute_Error"] = np.abs(

    prediction_df[
        "Actual_Required_Workers"
    ]

    - prediction_df[
        "PSO_Predicted_Required_Workers"
    ]
)

prediction_df["Percentage_Error"] = (

    prediction_df["Absolute_Error"]

    / np.maximum(

        prediction_df[
            "Actual_Required_Workers"
        ],

        1
    )

) * 100

# ==============================================================
# 19. WORKFORCE REQUIREMENT ANALYSIS
# ==============================================================

prediction_df["PSO_Workforce_Shortage"] = (
    np.maximum(

        prediction_df[
            "PSO_Predicted_Required_Workers"
        ]

        - prediction_df[
            "available_workers"
        ],

        0
    )
)

prediction_df["PSO_Workforce_Surplus"] = (
    np.maximum(

        prediction_df[
            "available_workers"
        ]

        - prediction_df[
            "PSO_Predicted_Required_Workers"
        ],

        0
    )
)

prediction_df["PSO_Workforce_Utilization"] = (

    np.minimum(

        prediction_df[
            "PSO_Predicted_Required_Workers"
        ]

        / np.maximum(

            prediction_df[
                "available_workers"
            ],

            1
        ),

        1.0
    )

    * 100
)

prediction_df["PSO_Estimated_Daily_Cost"] = (

    prediction_df[
        "PSO_Predicted_Required_Workers"
    ]

    * prediction_df[
        "daily_wage"
    ]
)

prediction_df["PSO_Additional_Workforce_Cost"] = (

    prediction_df[
        "PSO_Workforce_Shortage"
    ]

    * prediction_df[
        "daily_wage"
    ]
)

prediction_df["PSO_Workforce_Status"] = np.select(

    [

        prediction_df[
            "PSO_Workforce_Shortage"
        ] > 0,

        prediction_df[
            "PSO_Workforce_Surplus"
        ] > 0
    ],

    [

        "Workforce Shortage",

        "Workforce Surplus"
    ],

    default="Balanced Workforce"
)

prediction_df["PSO_Recommendation"] = np.select(

    [

        prediction_df[
            "PSO_Workforce_Shortage"
        ] > 0,

        prediction_df[
            "PSO_Workforce_Surplus"
        ] > 0
    ],

    [

        "Allocate Additional Workers",

        "Consider Workforce Redistribution"
    ],

    default="Maintain Current Workforce"
)

prediction_df["Data_Type"] = (
    "Synthetic Simulation"
)

prediction_df.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "pso_prediction.csv"
    ),

    index=False
)

print("\nPSO Prediction Preview:")

print(

    prediction_df[[

        "Scenario_ID",

        "available_workers",

        "Actual_Required_Workers",

        "PSO_Predicted_Required_Workers",

        "Absolute_Error",

        "PSO_Workforce_Status"

    ]].head(10).to_string(index=False)
)

# ==============================================================
# 20. GRAPH SAVE AND DISPLAY FUNCTION
# ==============================================================

def save_show(filename):

    if not filename.startswith("pso_"):

        filename = "pso_" + filename

    plt.tight_layout()

    plt.savefig(

        os.path.join(
            OUTPUT_DIR,
            filename
        ),

        dpi=300,

        bbox_inches="tight",

        facecolor="white"
    )

    plt.show()

    plt.close()

    print("Saved:", filename)

# ==============================================================
# 21. PSO ACCURACY GRAPH
# ==============================================================

plt.figure(figsize=(10, 6))

labels = [

    "Training",

    "Validation",

    "Testing"
]

scores = [

    train_metrics["R2_Percentage"],

    validation_metrics["R2_Percentage"],

    test_metrics["R2_Percentage"]
]

bars = plt.bar(

    labels,

    scores,

    color=[

        "#3498db",

        "#e67e22",

        "#2ecc71"
    ],

    edgecolor="black",

    width=0.55
)

for bar, value in zip(bars, scores):

    plt.annotate(

        f"{value:.2f}%",

        (

            bar.get_x()
            + bar.get_width() / 2,

            value
        ),

        ha="center",

        va="bottom" if value >= 0 else "top",

        xytext=(0, 5 if value >= 0 else -5),

        textcoords="offset points",

        fontweight="bold"
    )

plt.ylim(

    min(0, min(scores) - 10),

    max(110, max(scores) + 10)
)

plt.axhline(
    0,
    color="black"
)

plt.title(

    "PSO-Based Municipal Workforce Prediction - R² Performance",

    fontsize=15,

    fontweight="bold"
)

plt.xlabel("Dataset Split")

plt.ylabel("R² Score (%)")

save_show("pso_accuracy_graph.png")

# ==============================================================
# 22. PSO CORRELATION HEATMAP
# ==============================================================

correlation_matrix = df[

    FEATURES + [TARGET]

].corr()

plt.figure(figsize=(15, 11))

sns.heatmap(

    correlation_matrix,

    annot=True,

    fmt=".2f",

    cmap="coolwarm",

    center=0,

    vmin=-1,

    vmax=1,

    linewidths=0.5
)

plt.title(

    "PSO Municipal Workforce - Feature Correlation Heatmap",

    fontsize=16,

    fontweight="bold"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.yticks(rotation=0)

save_show("pso_heatmap.png")

correlation_matrix.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "pso_correlation_matrix.csv"
    )
)

# ==============================================================
# 23. PSO VS AIS VS BASELINE COMPARISON
# ==============================================================

comparison_columns = [

    "Model",

    "MAE",

    "MSE",

    "RMSE",

    "R2_Score",

    "R2_Percentage",

    "MAPE_Percentage"
]

comparison_frames = []

# Load conventional ML model results.

baseline_path = os.path.join(

    OUTPUT_DIR,

    "result.csv"
)

if os.path.exists(baseline_path):

    baseline_df = pd.read_csv(
        baseline_path
    )

    if "Dataset" in baseline_df.columns:

        baseline_df = baseline_df[

            baseline_df["Dataset"]
            .astype(str)
            .str.lower()
            .isin(["test", "testing"])
        ]

    if set(comparison_columns).issubset(
        baseline_df.columns
    ):

        comparison_frames.append(

            baseline_df[
                comparison_columns
            ]
        )

    else:

        print(
            "Baseline comparison columns do not match."
        )

else:

    print("Baseline result.csv not found.")

# Load AIS testing results.

ais_path = os.path.join(

    OUTPUT_DIR,

    "ais_result.csv"
)

if os.path.exists(ais_path):

    ais_df = pd.read_csv(
        ais_path
    )

    if "Dataset" in ais_df.columns:

        ais_df = ais_df[

            ais_df["Dataset"]
            .astype(str)
            .str.lower()
            .isin(["test", "testing"])
        ]

    if set(comparison_columns).issubset(
        ais_df.columns
    ):

        comparison_frames.append(

            ais_df[
                comparison_columns
            ]
        )

    else:

        print(
            "AIS comparison columns do not match."
        )

else:

    print("AIS result file not found.")

# Add PSO testing results.

pso_comparison_row = pd.DataFrame([{

    "Model": "PSO-ANN",

    **test_metrics
}])[comparison_columns]

comparison_frames.append(
    pso_comparison_row
)

comparison_df = pd.concat(

    comparison_frames,

    ignore_index=True
)

comparison_df = comparison_df.drop_duplicates(

    subset="Model",

    keep="last"
)

comparison_df = comparison_df.sort_values(

    "R2_Score",

    ascending=False

).reset_index(drop=True)

comparison_df.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "pso_comparison.csv"
    ),

    index=False
)

print("\n" + "=" * 75)
print("PSO VS AIS VS BASELINE MODEL COMPARISON")
print("=" * 75)

print(
    comparison_df.round(4).to_string(
        index=False
    )
)

# ==============================================================
# 24. PSO COMPARISON GRAPH
# ==============================================================

fig, axes = plt.subplots(

    2,

    2,

    figsize=(17, 12)
)

comparison_metrics = [

    ("R2_Percentage", "R² Score (%)"),

    ("MAE", "Mean Absolute Error"),

    ("RMSE", "Root Mean Squared Error"),

    ("MAPE_Percentage", "MAPE (%)")
]

colors = [

    "#e74c3c" if name == "PSO-ANN"

    else "#2ecc71" if name == "AIS-ANN"

    else "#3498db"

    for name in comparison_df["Model"]
]

for ax, (metric, title) in zip(

    axes.flatten(),

    comparison_metrics
):

    bars = ax.bar(

        comparison_df["Model"],

        comparison_df[metric],

        color=colors,

        edgecolor="black"
    )

    ax.set_title(

        title,

        fontweight="bold"
    )

    ax.tick_params(

        axis="x",

        rotation=25
    )

    ax.margins(y=0.18)

    for bar, value in zip(

        bars,

        comparison_df[metric]
    ):

        ax.annotate(

            f"{value:.2f}",

            (

                bar.get_x()
                + bar.get_width() / 2,

                value
            ),

            ha="center",

            va="bottom" if value >= 0 else "top",

            fontsize=9
        )

fig.suptitle(

    "PSO vs AIS vs Conventional Machine Learning Models",

    fontsize=18,

    fontweight="bold"
)

save_show("pso_comparison_graph.png")

# ==============================================================
# 25. PSO RESULT GRAPH
# ==============================================================

fig, axes = plt.subplots(

    2,

    2,

    figsize=(13, 10)
)

metrics_to_plot = [

    ("MAE", "Mean Absolute Error"),

    ("RMSE", "Root Mean Squared Error"),

    ("R2_Score", "R² Score"),

    ("MAPE_Percentage", "MAPE (%)")
]

for ax, (metric, title) in zip(

    axes.flatten(),

    metrics_to_plot
):

    bars = ax.bar(

        result_df["Dataset"],

        result_df[metric],

        color=[

            "#3498db",

            "#e67e22",

            "#2ecc71"
        ],

        edgecolor="black"
    )

    ax.set_title(

        title,

        fontweight="bold"
    )

    ax.margins(y=0.18)

    for bar, value in zip(

        bars,

        result_df[metric]
    ):

        ax.annotate(

            f"{value:.3f}",

            (

                bar.get_x()
                + bar.get_width() / 2,

                value
            ),

            ha="center",

            va="bottom" if value >= 0 else "top"
        )

fig.suptitle(

    "PSO Training, Validation and Testing Results",

    fontsize=17,

    fontweight="bold"
)

save_show("pso_result_graph.png")

# ==============================================================
# 26. PSO PREDICTION GRAPH
# ==============================================================

N_DISPLAY = min(

    100,

    len(prediction_df)
)

sample = prediction_df.head(
    N_DISPLAY
)

scenario_positions = np.arange(

    1,

    N_DISPLAY + 1
)

plt.figure(figsize=(16, 7))

plt.plot(

    scenario_positions,

    sample[
        "Actual_Required_Workers"
    ],

    label="Simulated Actual Workforce",

    color="#3498db",

    linewidth=2,

    marker="o",

    markersize=3
)

plt.plot(

    scenario_positions,

    sample[
        "PSO_Predicted_Required_Workers"
    ],

    label="PSO Predicted Workforce",

    color="#e74c3c",

    linestyle="--",

    linewidth=2,

    marker="x",

    markersize=3
)

plt.title(

    "PSO - Actual vs Predicted Municipal Workforce",

    fontsize=16,

    fontweight="bold"
)

plt.xlabel("Test Scenario")

plt.ylabel("Required Workers")

plt.legend()

save_show("pso_prediction_graph.png")

# ==============================================================
# 27. PSO CONVERGENCE GRAPH
# ==============================================================

plt.figure(figsize=(13, 7))

plt.plot(

    history_df["Iteration"],

    history_df["Training_Loss"],

    label="Training Loss",

    color="#3498db",

    linewidth=2
)

plt.plot(

    history_df["Iteration"],

    history_df["Validation_Loss"],

    label="Validation Loss",

    color="#e67e22",

    alpha=0.7
)

plt.plot(

    history_df["Iteration"],

    history_df["Best_Validation_Loss"],

    label="Best Validation Loss",

    color="#2ecc71",

    linewidth=2.5
)

plt.axvline(

    best_iteration,

    color="red",

    linestyle="--",

    label=f"Best Iteration: {best_iteration}"
)

plt.title(

    "Particle Swarm Optimization - Fitness Convergence",

    fontsize=16,

    fontweight="bold"
)

plt.xlabel("Iteration")

plt.ylabel("Scaled Mean Squared Error")

plt.legend()

save_show("pso_convergence_graph.png")

# ==============================================================
# 28. PSO FITNESS GRAPH
# ==============================================================

plt.figure(figsize=(13, 7))

plt.plot(

    history_df["Iteration"],

    history_df["Global_Best_Fitness"],

    color="#2ecc71",

    linewidth=2,

    label="Global Best Fitness"
)

plt.plot(

    history_df["Iteration"],

    history_df["Average_Swarm_Fitness"],

    color="#3498db",

    linewidth=2,

    alpha=0.7,

    label="Average Swarm Fitness"
)

plt.title(

    "PSO Global Best and Average Swarm Fitness",

    fontsize=16,

    fontweight="bold"
)

plt.xlabel("Iteration")

plt.ylabel("Regularized Fitness")

plt.legend()

save_show("pso_fitness_graph.png")

# ==============================================================
# 29. PSO SWARM DIVERSITY GRAPH
# ==============================================================

plt.figure(figsize=(12, 7))

plt.plot(

    history_df["Iteration"],

    history_df["Swarm_Diversity"],

    color="#9b59b6",

    linewidth=2
)

plt.title(

    "PSO Particle Swarm Diversity",

    fontsize=16,

    fontweight="bold"
)

plt.xlabel("Iteration")

plt.ylabel("Mean Parameter Standard Deviation")

save_show("pso_swarm_diversity_graph.png")

# ==============================================================
# 30. PSO VELOCITY GRAPH
# ==============================================================

plt.figure(figsize=(12, 7))

plt.plot(

    history_df["Iteration"],

    history_df["Average_Velocity"],

    color="#e67e22",

    linewidth=2
)

plt.title(

    "PSO Average Particle Velocity",

    fontsize=16,

    fontweight="bold"
)

plt.xlabel("Iteration")

plt.ylabel("Mean Absolute Velocity")

save_show("pso_velocity_graph.png")

# ==============================================================
# 31. PSO INERTIA WEIGHT GRAPH
# ==============================================================

plt.figure(figsize=(12, 7))

plt.plot(

    history_df["Iteration"],

    history_df["Inertia_Weight"],

    color="#16a085",

    linewidth=2
)

plt.title(

    "PSO Adaptive Inertia Weight",

    fontsize=16,

    fontweight="bold"
)

plt.xlabel("Iteration")

plt.ylabel("Inertia Weight")

save_show("pso_inertia_graph.png")

# ==============================================================
# 32. PSO ACTUAL VS PREDICTED SCATTER GRAPH
# ==============================================================

actual = prediction_df[
    "Actual_Required_Workers"
]

predicted = prediction_df[
    "PSO_Predicted_Required_Workers"
]

plt.figure(figsize=(9, 8))

plt.scatter(

    actual,

    predicted,

    alpha=0.5,

    color="#3498db"
)

min_value = min(

    actual.min(),

    predicted.min()
)

max_value = max(

    actual.max(),

    predicted.max()
)

plt.plot(

    [min_value, max_value],

    [min_value, max_value],

    color="red",

    linestyle="--",

    label="Perfect Prediction"
)

plt.title(

    f"PSO Actual vs Predicted Workforce\n"

    f"R² = {test_metrics['R2_Score']:.4f}",

    fontsize=15,

    fontweight="bold"
)

plt.xlabel("Simulated Actual Workforce")

plt.ylabel("PSO Predicted Workforce")

plt.legend()

save_show("pso_actual_vs_predicted.png")

# ==============================================================
# 33. PSO RESIDUAL ERROR GRAPH
# ==============================================================

residuals = actual - predicted

plt.figure(figsize=(12, 7))

plt.scatter(

    predicted,

    residuals,

    alpha=0.5,

    color="#9b59b6"
)

plt.axhline(

    0,

    color="red",

    linestyle="--"
)

plt.title(

    "PSO Workforce Prediction - Residual Analysis",

    fontsize=16,

    fontweight="bold"
)

plt.xlabel("PSO Predicted Workforce")

plt.ylabel("Residual Error")

save_show("pso_residual_graph.png")

# ==============================================================
# 34. PSO ERROR DISTRIBUTION
# ==============================================================

plt.figure(figsize=(12, 7))

sns.histplot(

    residuals,

    bins=40,

    kde=True,

    color="#3498db"
)

plt.axvline(

    0,

    color="red",

    linestyle="--"
)

plt.title(

    "PSO Prediction Error Distribution",

    fontsize=16,

    fontweight="bold"
)

plt.xlabel("Prediction Error")

plt.ylabel("Frequency")

save_show("pso_error_distribution_graph.png")

# ==============================================================
# 35. PSO WORKFORCE ALLOCATION GRAPH
# ==============================================================

allocation_sample = prediction_df.head(25)

x_positions = np.arange(
    len(allocation_sample)
)

width = 0.4

plt.figure(figsize=(16, 7))

plt.bar(

    x_positions - width / 2,

    allocation_sample[
        "available_workers"
    ],

    width,

    color="#3498db",

    label="Available Workforce"
)

plt.bar(

    x_positions + width / 2,

    allocation_sample[
        "PSO_Predicted_Required_Workers"
    ],

    width,

    color="#e67e22",

    label="PSO Required Workforce"
)

plt.xticks(

    x_positions,

    allocation_sample["Scenario_ID"],

    rotation=45
)

plt.title(

    "PSO-Based Municipal Workforce Requirement Analysis",

    fontsize=16,

    fontweight="bold"
)

plt.xlabel("Scenario ID")

plt.ylabel("Number of Workers")

plt.legend()

save_show("pso_workforce_allocation_graph.png")

# ==============================================================
# 36. PSO SHORTAGE AND SURPLUS GRAPH
# ==============================================================

plt.figure(figsize=(16, 7))

plt.bar(

    x_positions - width / 2,

    allocation_sample[
        "PSO_Workforce_Shortage"
    ],

    width,

    color="#e74c3c",

    label="Workforce Shortage"
)

plt.bar(

    x_positions + width / 2,

    allocation_sample[
        "PSO_Workforce_Surplus"
    ],

    width,

    color="#2ecc71",

    label="Workforce Surplus"
)

plt.xticks(

    x_positions,

    allocation_sample["Scenario_ID"],

    rotation=45
)

plt.title(

    "PSO Workforce Shortage and Surplus Analysis",

    fontsize=16,

    fontweight="bold"
)

plt.xlabel("Scenario ID")

plt.ylabel("Number of Workers")

plt.legend()

save_show("pso_workforce_shortage_graph.png")

# ==============================================================
# 37. PSO WORKFORCE STATUS GRAPH
# ==============================================================

status_counts = prediction_df[

    "PSO_Workforce_Status"

].value_counts()

plt.figure(figsize=(9, 8))

status_colors = {

    "Workforce Shortage": "#e74c3c",

    "Workforce Surplus": "#2ecc71",

    "Balanced Workforce": "#3498db"
}

plt.pie(

    status_counts.values,

    labels=status_counts.index,

    autopct="%1.1f%%",

    startangle=90,

    colors=[

        status_colors[name]

        for name in status_counts.index
    ],

    wedgeprops={

        "edgecolor": "white",

        "linewidth": 2
    }
)

plt.title(

    "PSO Predicted Workforce Status Distribution",

    fontsize=16,

    fontweight="bold"
)

save_show("pso_workforce_status_graph.png")

# ==============================================================
# 38. PSO WORKFORCE UTILIZATION GRAPH
# ==============================================================

plt.figure(figsize=(12, 7))

sns.histplot(

    prediction_df[
        "PSO_Workforce_Utilization"
    ],

    bins=25,

    color="#16a085"
)

average_utilization = prediction_df[

    "PSO_Workforce_Utilization"

].mean()

plt.axvline(

    average_utilization,

    color="red",

    linestyle="--",

    label=f"Average: {average_utilization:.2f}%"
)

plt.title(

    "PSO Municipal Workforce Utilization",

    fontsize=16,

    fontweight="bold"
)

plt.xlabel("Workforce Utilization (%)")

plt.ylabel("Number of Scenarios")

plt.legend()

save_show("pso_workforce_utilization_graph.png")

# ==============================================================
# 39. PSO OPERATIONAL COST GRAPH
# ==============================================================

actual_daily_cost = (

    sample[
        "Actual_Required_Workers"
    ]

    * sample[
        "daily_wage"
    ]
)

plt.figure(figsize=(15, 7))

plt.plot(

    scenario_positions,

    actual_daily_cost,

    color="#3498db",

    label="Simulated Reference Cost",

    linewidth=2
)

plt.plot(

    scenario_positions,

    sample[
        "PSO_Estimated_Daily_Cost"
    ],

    color="#e67e22",

    linestyle="--",

    label="PSO Estimated Cost",

    linewidth=2
)

plt.title(

    "PSO Municipal Workforce - Operational Cost Analysis",

    fontsize=16,

    fontweight="bold"
)

plt.xlabel("Test Scenario")

plt.ylabel("Estimated Daily Cost (INR)")

plt.legend()

save_show("pso_operational_cost_graph.png")

# ==============================================================
# 40. PSO PERMUTATION FEATURE IMPORTANCE
# ==============================================================

baseline_mae = test_metrics["MAE"]

importance_results = []

for feature in FEATURES:

    repeated_importance = []

    for repeat in range(5):

        X_permuted = X_test.copy()

        X_permuted[feature] = rng.permutation(

            X_permuted[feature].to_numpy()
        )

        X_permuted_scaled = (
            feature_scaler.transform(
                X_permuted
            ).astype(np.float32)
        )

        permuted_prediction = predict_workforce(

            X_permuted_scaled,

            X_permuted
        )

        permuted_mae = mean_absolute_error(

            y_test,

            permuted_prediction
        )

        repeated_importance.append(

            permuted_mae - baseline_mae
        )

    importance_results.append({

        "Feature": feature,

        "Importance": float(
            np.mean(repeated_importance)
        ),

        "Standard_Deviation": float(
            np.std(repeated_importance)
        )
    })

importance_df = pd.DataFrame(
    importance_results
)

importance_df = importance_df.sort_values(

    "Importance",

    ascending=True
)

importance_df.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "pso_feature_importance.csv"
    ),

    index=False
)

plt.figure(figsize=(12, 8))

plt.barh(

    importance_df["Feature"],

    importance_df["Importance"],

    xerr=importance_df[
        "Standard_Deviation"
    ],

    color="#2ecc71",

    edgecolor="black"
)

plt.title(

    "PSO Permutation Feature Importance",

    fontsize=16,

    fontweight="bold"
)

plt.xlabel("Increase in MAE After Permutation")

plt.axvline(

    0,

    color="black"
)

save_show("pso_feature_importance_graph.png")

# ==============================================================
# 41. DIRECT PSO VS AIS PREDICTION COMPARISON
# ==============================================================

ais_prediction_path = os.path.join(

    OUTPUT_DIR,

    "ais_prediction.csv"
)

if os.path.exists(ais_prediction_path):

    ais_predictions = pd.read_csv(
        ais_prediction_path
    )

    required_ais_columns = [

        "Scenario_ID",

        "Actual_Required_Workers",

        "AIS_Predicted_Required_Workers"
    ]

    if set(required_ais_columns).issubset(
        ais_predictions.columns
    ):

        direct_comparison = prediction_df[[

            "Scenario_ID",

            "Actual_Required_Workers",

            "PSO_Predicted_Required_Workers"

        ]].merge(

            ais_predictions[
                required_ais_columns
            ],

            on="Scenario_ID",

            suffixes=("", "_AIS"),

            validate="one_to_one"
        )

        # Confirm that reference targets match.

        targets_match = np.allclose(

            direct_comparison[
                "Actual_Required_Workers"
            ],

            direct_comparison[
                "Actual_Required_Workers_AIS"
            ]
        )

        complete_match = (

            len(direct_comparison)

            == len(prediction_df)

            == len(ais_predictions)
        )

        if targets_match and complete_match:

            direct_comparison = (
                direct_comparison.drop(
                    columns=[
                        "Actual_Required_Workers_AIS"
                    ]
                )
            )

            direct_comparison["AIS_Absolute_Error"] = (

                np.abs(

                    direct_comparison[
                        "Actual_Required_Workers"
                    ]

                    - direct_comparison[
                        "AIS_Predicted_Required_Workers"
                    ]
                )
            )

            direct_comparison["PSO_Absolute_Error"] = (

                np.abs(

                    direct_comparison[
                        "Actual_Required_Workers"
                    ]

                    - direct_comparison[
                        "PSO_Predicted_Required_Workers"
                    ]
                )
            )

            direct_comparison.to_csv(

                os.path.join(

                    OUTPUT_DIR,

                    "pso_ais_prediction_comparison.csv"
                ),

                index=False
            )

            direct_sample = (
                direct_comparison.head(100)
            )

            x_direct = np.arange(

                1,

                len(direct_sample) + 1
            )

            plt.figure(figsize=(16, 7))

            plt.plot(

                x_direct,

                direct_sample[
                    "Actual_Required_Workers"
                ],

                label="Simulated Actual",

                color="#3498db",

                linewidth=2
            )

            plt.plot(

                x_direct,

                direct_sample[
                    "AIS_Predicted_Required_Workers"
                ],

                label="AIS Prediction",

                color="#2ecc71",

                linestyle="--"
            )

            plt.plot(

                x_direct,

                direct_sample[
                    "PSO_Predicted_Required_Workers"
                ],

                label="PSO Prediction",

                color="#e74c3c",

                linestyle=":"
            )

            plt.title(

                "PSO vs AIS - Workforce Prediction Comparison",

                fontsize=16,

                fontweight="bold"
            )

            plt.xlabel("Test Scenario")

            plt.ylabel("Required Workers")

            plt.legend()

            save_show(

                "pso_ais_prediction_comparison_graph.png"
            )

        else:

            print(
                "AIS and PSO test scenarios do not match. "
                "Direct prediction comparison skipped."
            )

# ==============================================================
# 42. PSO VS AIS CONVERGENCE COMPARISON
# ==============================================================

ais_history_path = os.path.join(

    OUTPUT_DIR,

    "ais_convergence_history.csv"
)

if os.path.exists(ais_history_path):

    ais_history = pd.read_csv(
        ais_history_path
    )

    if {

        "Generation",

        "Best_Validation_Loss"

    }.issubset(ais_history.columns):

        plt.figure(figsize=(13, 7))

        plt.plot(

            ais_history["Generation"],

            ais_history[
                "Best_Validation_Loss"
            ],

            label="AIS",

            color="#2ecc71",

            linewidth=2
        )

        plt.plot(

            history_df["Iteration"],

            history_df[
                "Best_Validation_Loss"
            ],

            label="PSO",

            color="#e74c3c",

            linewidth=2
        )

        plt.title(

            "AIS vs PSO - Validation Loss Convergence",

            fontsize=16,

            fontweight="bold"
        )

        plt.xlabel("Generation / Iteration")

        plt.ylabel("Best Validation MSE")

        plt.legend()

        save_show(

            "pso_ais_convergence_comparison.png"
        )

# ==============================================================
# 43. PSO WORKFORCE SUMMARY
# ==============================================================

summary = {

    "Algorithm": "Particle Swarm Optimization",

    "Total_Test_Scenarios": int(
        len(prediction_df)
    ),

    "Best_Iteration": int(
        best_iteration
    ),

    "Global_Best_Fitness": float(
        global_best_score
    ),

    "Best_Validation_Loss": float(
        best_validation_loss
    ),

    "Fitness_Evaluations": int(
        fitness_evaluations
    ),

    **test_metrics,

    "Average_Available_Workforce": float(

        prediction_df[
            "available_workers"
        ].mean()
    ),

    "Average_Predicted_Workforce": float(

        prediction_df[
            "PSO_Predicted_Required_Workers"
        ].mean()
    ),

    "Average_Workforce_Shortage": float(

        prediction_df[
            "PSO_Workforce_Shortage"
        ].mean()
    ),

    "Average_Workforce_Surplus": float(

        prediction_df[
            "PSO_Workforce_Surplus"
        ].mean()
    ),

    "Average_Workforce_Utilization": float(

        prediction_df[
            "PSO_Workforce_Utilization"
        ].mean()
    ),

    "Average_Daily_Cost": float(

        prediction_df[
            "PSO_Estimated_Daily_Cost"
        ].mean()
    ),

    "Shortage_Scenarios": int(

        (

            prediction_df[
                "PSO_Workforce_Status"
            ]

            == "Workforce Shortage"

        ).sum()
    ),

    "Surplus_Scenarios": int(

        (

            prediction_df[
                "PSO_Workforce_Status"
            ]

            == "Workforce Surplus"

        ).sum()
    ),

    "Balanced_Scenarios": int(

        (

            prediction_df[
                "PSO_Workforce_Status"
            ]

            == "Balanced Workforce"

        ).sum()
    ),

    "Optimization_Time_Seconds": float(
        optimization_time
    ),

    "Execution_Time_Seconds": float(
        time.time() - START_TIME
    ),

    "Evaluation_Type": "Synthetic Simulation",

    "Real_World_Validated": False
}

pd.DataFrame([summary]).to_csv(

    os.path.join(
        OUTPUT_DIR,
        "pso_workforce_summary.csv"
    ),

    index=False
)

# ==============================================================
# 44. SAVE PSO YAML CONFIGURATION
# ==============================================================

configuration = {

    "project": {

        "name": (

            "PSO-Based Municipal Waste Management "

            "and Conservancy Workforce Optimization"
        ),

        "version": "1.0",

        "algorithm": "Particle Swarm Optimization",

        "random_seed": SEED,

        "data_type": "synthetic_simulation"
    },

    "dataset": {

        "original_dataset": ORIGINAL_DATASET,

        "synthetic_dataset": SYNTHETIC_DATASET,

        "total_records": int(len(df)),

        "training_records": int(len(X_train)),

        "validation_records": int(len(X_val)),

        "testing_records": int(len(X_test)),

        "features": FEATURES,

        "target": TARGET
    },

    "pso_parameters": {

        "swarm_size": SWARM_SIZE,

        "iterations": ITERATIONS,

        "maximum_inertia_weight": W_MAX,

        "minimum_inertia_weight": W_MIN,

        "cognitive_coefficient": C1,

        "social_coefficient": C2,

        "position_limit": POSITION_LIMIT,

        "velocity_limit": VELOCITY_LIMIT,

        "regularization": REGULARIZATION,

        "fitness_function": (

            "Training MSE with weight regularization"
        ),

        "checkpoint_selection": (

            "Minimum validation MSE"
        ),

        "fitness_evaluations": int(
            fitness_evaluations
        )
    },

    "neural_network": {

        "input_size": INPUT_SIZE,

        "hidden_layer_1": HIDDEN_LAYER_1,

        "hidden_layer_2": HIDDEN_LAYER_2,

        "output_size": 1,

        "activation": "ReLU",

        "total_parameters": TOTAL_PARAMETERS,

        "gradient_training": False
    },

    "constraints": {

        "minimum_workers_per_site": 2,

        "integer_workforce_predictions": True
    },

    "files": {

        "h5_model": "pso_model.h5",

        "pkl_model": "pso_model.pkl",

        "feature_scaler": "pso_scaler.pkl",

        "target_scaler": "pso_target_scaler.pkl",

        "best_particle": "pso_best_particle.npy",

        "results": "pso_result.csv",

        "predictions": "pso_prediction.csv"
    },

    "limitations": [

        "Original dataset contains aggregate statistics.",

        "Training and testing use synthetic scenarios.",

        "PSO optimizes prediction model parameters.",

        "Site-level workforce allocation is not validated.",

        "AIS and PSO use different fitness evaluation budgets.",

        "Real-world municipal performance is not established."
    ]
}

with open(

    os.path.join(
        OUTPUT_DIR,
        "pso_config.yaml"
    ),

    "w",

    encoding="utf-8"

) as file:

    yaml.safe_dump(

        configuration,

        file,

        sort_keys=False,

        allow_unicode=True
    )

print("\nPSO YAML Configuration Saved.")

# ==============================================================
# 45. SAVE PSO JSON METRICS
# ==============================================================

metrics_report = {

    "project": (

        "PSO-Based Municipal Waste Management "

        "and Conservancy Workforce Optimization"
    ),

    "algorithm": "Particle Swarm Optimization",

    "evaluation_type": "Synthetic Simulation",

    "best_iteration": best_iteration,

    "best_validation_loss": float(
        best_validation_loss
    ),

    "global_best_fitness": float(
        global_best_score
    ),

    "fitness_evaluations": int(
        fitness_evaluations
    ),

    "training_metrics": train_metrics,

    "validation_metrics": validation_metrics,

    "test_metrics": test_metrics,

    "model_comparison": comparison_df.to_dict(
        orient="records"
    ),

    "workforce_summary": summary,

    "real_world_validated": False
}

with open(

    os.path.join(
        OUTPUT_DIR,
        "pso_metrics.json"
    ),

    "w",

    encoding="utf-8"

) as file:

    json.dump(

        metrics_report,

        file,

        indent=4,

        allow_nan=False
    )

# Save neural network architecture.

with open(

    os.path.join(
        OUTPUT_DIR,
        "pso_model_architecture.json"
    ),

    "w",

    encoding="utf-8"

) as file:

    file.write(
        pso_model.to_json()
    )

print("PSO JSON Metrics Saved.")

# ==============================================================
# 46. PREDICT NEW MUNICIPAL SCENARIO
# ==============================================================

new_scenario = pd.DataFrame([{

    "disposal_sites": 105,

    "other_workers": 412,

    "daily_waste_tonnes": 650,

    "population_density": 12000,

    "working_hours": 8,

    "worker_productivity": 1.0,

    "equipment_efficiency": 0.85,

    "absenteeism_rate": 0.08,

    "transport_distance_km": 10,

    "seasonal_demand_factor": 1.1
}])

new_scaled = feature_scaler.transform(

    new_scenario[FEATURES]

).astype(np.float32)

new_prediction = int(

    predict_workforce(

        new_scaled,

        new_scenario

    )[0]
)

available_workers = 847

shortage = max(

    new_prediction - available_workers,

    0
)

surplus = max(

    available_workers - new_prediction,

    0
)

new_scenario["Available_Workers"] = (
    available_workers
)

new_scenario["PSO_Predicted_Workers"] = (
    new_prediction
)

new_scenario["PSO_Workforce_Shortage"] = (
    shortage
)

new_scenario["PSO_Workforce_Surplus"] = (
    surplus
)

new_scenario["Data_Type"] = (
    "Synthetic Simulation"
)

new_scenario.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "pso_new_scenario_prediction.csv"
    ),

    index=False
)

print("\n" + "=" * 75)
print("PSO NEW MUNICIPAL SCENARIO")
print("=" * 75)

print("Available Workers:", available_workers)

print("Predicted Workers:", new_prediction)

print("Workforce Shortage:", shortage)

print("Workforce Surplus:", surplus)

# ==============================================================
# 47. DISPLAY FINAL SUMMARY
# ==============================================================

print("\n" + "=" * 75)
print("FINAL PSO PERFORMANCE SUMMARY")
print("=" * 75)

for key, value in summary.items():

    if isinstance(value, float):

        print(

            f"{key:<40}: {value:.4f}"
        )

    else:

        print(

            f"{key:<40}: {value}"
        )

# ==============================================================
# 48. DISPLAY GENERATED PSO FILES
# ==============================================================

print("\n" + "=" * 75)
print("GENERATED PSO OUTPUT FILES")
print("=" * 75)

file_count = 0

for filename in sorted(

    os.listdir(OUTPUT_DIR)
):

    if filename.startswith("pso_"):

        filepath = os.path.join(

            OUTPUT_DIR,

            filename
        )

        if os.path.isfile(filepath):

            size_kb = (

                os.path.getsize(filepath)

                / 1024
            )

            print(

                f"{filename:<50} "

                f"{size_kb:>10.2f} KB"
            )

            file_count += 1

print("\nTotal PSO Files:", file_count)

print("\nOutput Directory:")

print(OUTPUT_DIR)

print("\nTotal Execution Time:")

print(

    f"{time.time() - START_TIME:.2f} seconds"
)

print("\nPSO PROJECT COMPLETED SUCCESSFULLY.")

print(

    "\nIMPORTANT: All reported results are based "

    "on synthetic operational scenarios and "

    "are not validated real-world municipal predictions."
)
```

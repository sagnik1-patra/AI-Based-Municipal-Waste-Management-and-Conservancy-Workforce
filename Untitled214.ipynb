```python
import os
import json
import yaml
import joblib
import warnings
import numpy as np
import pandas as pd
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import (
    RandomForestRegressor,
    GradientBoostingRegressor,
    ExtraTreesRegressor
)
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore")

# ==============================================================
# 1. CONFIGURATION
# ==============================================================

SEED = 42

np.random.seed(SEED)
tf.random.set_seed(SEED)

rng = np.random.default_rng(SEED)

DATASET_PATH = r"C:\Users\sagni\Downloads\AI-Based Municipal Waste Management and Conservancy Workforce\number_of conservancy_workers_engaged_in_conservancy_garbage_and_debris_disposal_and_other_particulars_2017_18.csv"

OUTPUT_DIR = r"C:\Users\sagni\Downloads\AI-Based Municipal Waste Management and Conservancy Workforce"

os.makedirs(OUTPUT_DIR, exist_ok=True)

N_SAMPLES = 10000

print("=" * 70)
print("AI-BASED MUNICIPAL WASTE MANAGEMENT")
print("=" * 70)

# ==============================================================
# 2. LOAD ORIGINAL DATASET
# ==============================================================

if not os.path.exists(DATASET_PATH):
    raise FileNotFoundError(
        f"Dataset not found: {DATASET_PATH}"
    )

df = pd.read_csv(DATASET_PATH)

print("\nOriginal Dataset:")
print(df)

print("\nDataset Shape:", df.shape)

df.columns = (
    df.columns.str.strip()
    .str.lower()
    .str.replace(r"[^a-z0-9]+", "_", regex=True)
    .str.strip("_")
)

print("\nDataset Columns:")
print(df.columns.tolist())

# ==============================================================
# 3. EXTRACT BASELINE PARAMETERS
# ==============================================================

worker_col = (
    "number_of_workers_engaged_in_conservancy_work"
)

site_col = (
    "number_of_garbage_and_sewage_disposal_dump_site"
)

other_worker_col = (
    "number_of_persons_engaged_in_other_connected_"
    "works_related_with_conservancy"
)

required_columns = [
    worker_col,
    site_col,
    other_worker_col
]

for col in required_columns:
    if col not in df.columns:
        raise ValueError(f"Missing column: {col}")

for col in required_columns:
    df[col] = pd.to_numeric(df[col], errors="raise")

BASE_WORKERS = int(df[worker_col].sum())
BASE_SITES = int(df[site_col].sum())
BASE_OTHER_WORKERS = int(df[other_worker_col].sum())

if min(BASE_WORKERS, BASE_SITES, BASE_OTHER_WORKERS) <= 0:
    raise ValueError("Baseline values must be positive.")

TOTAL_WORKFORCE = BASE_WORKERS + BASE_OTHER_WORKERS

WORKERS_PER_SITE = BASE_WORKERS / BASE_SITES

print("\nBaseline Statistics")
print("-" * 50)

print("Conservancy Workers:", BASE_WORKERS)
print("Disposal Sites:", BASE_SITES)
print("Other Workers:", BASE_OTHER_WORKERS)
print("Total Workforce:", TOTAL_WORKFORCE)
print("Workers Per Site:", round(WORKERS_PER_SITE, 2))

# ==============================================================
# 4. SYNTHETIC DATA GENERATION
# ==============================================================

# The original dataset contains only one aggregate record.
# Additional operational variables are simulated assumptions.
# These scenarios are not actual municipal observations.

print("\nGenerating Synthetic Operational Scenarios...")

disposal_sites = rng.integers(
    max(1, int(BASE_SITES * 0.60)),
    max(2, int(BASE_SITES * 1.50)),
    N_SAMPLES
)

available_workers = rng.integers(
    max(1, int(BASE_WORKERS * 0.60)),
    max(2, int(BASE_WORKERS * 1.50)),
    N_SAMPLES
)

other_workers = rng.integers(
    max(1, int(BASE_OTHER_WORKERS * 0.60)),
    max(2, int(BASE_OTHER_WORKERS * 1.50)),
    N_SAMPLES
)

# Assumed waste generation in tonnes per site per day
waste_per_site = rng.uniform(
    3.0, 12.0, N_SAMPLES
)

daily_waste = disposal_sites * waste_per_site

population_density = rng.uniform(
    2000, 25000, N_SAMPLES
)

working_hours = rng.uniform(
    6, 10, N_SAMPLES
)

worker_productivity = rng.uniform(
    0.6, 1.4, N_SAMPLES
)

equipment_efficiency = rng.uniform(
    0.55, 0.98, N_SAMPLES
)

absenteeism_rate = rng.uniform(
    0.02, 0.20, N_SAMPLES
)

transport_distance = rng.uniform(
    2, 30, N_SAMPLES
)

seasonal_demand_factor = rng.uniform(
    0.8, 1.5, N_SAMPLES
)

daily_wage = rng.uniform(
    400, 1000, N_SAMPLES
)

# ==============================================================
# 5. SIMULATED WORKFORCE REQUIREMENT MODEL
# ==============================================================

# Assumed nominal waste-handling capacity:
# 0.18 tonnes per worker per hour

nominal_capacity = 0.18

effective_capacity = (
    nominal_capacity
    * working_hours
    * worker_productivity
    * equipment_efficiency
    * (1 - absenteeism_rate)
)

adjusted_waste_demand = (
    daily_waste * seasonal_demand_factor
)

density_factor = (
    1 + population_density / 100000
)

transport_factor = (
    1 + transport_distance / 100
)

support_factor = (
    1 + 0.15 * other_workers / BASE_OTHER_WORKERS
)

required_workers = (
    adjusted_waste_demand
    * density_factor
    * transport_factor
    / (effective_capacity * support_factor)
)

# Minimum staffing requirement per disposal site
minimum_staffing = disposal_sites * 2

required_workers = np.maximum(
    required_workers,
    minimum_staffing
)

# Introduce simulated operational uncertainty
noise = rng.normal(
    0, 0.03, N_SAMPLES
)

required_workers = np.ceil(
    required_workers * (1 + noise)
).astype(int)

required_workers = np.maximum(
    required_workers,
    minimum_staffing
)

workforce_shortage = np.maximum(
    required_workers - available_workers,
    0
)

workforce_surplus = np.maximum(
    available_workers - required_workers,
    0
)

workforce_utilization = np.minimum(
    required_workers / available_workers,
    1.0
) * 100

estimated_daily_cost = (
    required_workers * daily_wage
)

# ==============================================================
# 6. CREATE SYNTHETIC DATAFRAME
# ==============================================================

synthetic_df = pd.DataFrame({

    "disposal_sites": disposal_sites,

    "available_workers": available_workers,

    "other_workers": other_workers,

    "daily_waste_tonnes": daily_waste,

    "population_density": population_density,

    "working_hours": working_hours,

    "worker_productivity": worker_productivity,

    "equipment_efficiency": equipment_efficiency,

    "absenteeism_rate": absenteeism_rate,

    "transport_distance_km": transport_distance,

    "seasonal_demand_factor": seasonal_demand_factor,

    "daily_wage": daily_wage,

    "required_workers": required_workers,

    "workforce_shortage": workforce_shortage,

    "workforce_surplus": workforce_surplus,

    "workforce_utilization": workforce_utilization,

    "estimated_daily_cost": estimated_daily_cost
})

synthetic_df.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "synthetic_municipal_dataset.csv"
    ),
    index=False
)

print("\nSynthetic Dataset Shape:", synthetic_df.shape)

print("\nSynthetic Dataset Preview:")
print(synthetic_df.head())

# ==============================================================
# 7. FEATURE AND TARGET SELECTION
# ==============================================================

FEATURES = [
    "disposal_sites",
    "other_workers",
    "daily_waste_tonnes",
    "population_density",
    "working_hours",
    "worker_productivity",
    "equipment_efficiency",
    "absenteeism_rate",
    "transport_distance_km",
    "seasonal_demand_factor"
]

TARGET = "required_workers"

# Available workforce and wages are used for downstream
# allocation and cost analysis, not demand prediction.

X = synthetic_df[FEATURES]
y = synthetic_df[TARGET]

print("\nInput Features:", FEATURES)
print("\nTarget Variable:", TARGET)

# ==============================================================
# 8. TRAIN, VALIDATION AND TEST SPLIT
# ==============================================================

X_train_full, X_test, y_train_full, y_test = (
    train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=SEED
    )
)

X_train, X_val, y_train, y_val = (
    train_test_split(
        X_train_full,
        y_train_full,
        test_size=0.20,
        random_state=SEED
    )
)

print("\nTraining Samples:", len(X_train))
print("Validation Samples:", len(X_val))
print("Testing Samples:", len(X_test))

# ==============================================================
# 9. FEATURE SCALING
# ==============================================================

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_val_scaled = scaler.transform(X_val)

X_test_scaled = scaler.transform(X_test)

joblib.dump(
    scaler,
    os.path.join(OUTPUT_DIR, "scaler.pkl")
)

# ==============================================================
# 10. PERFORMANCE EVALUATION FUNCTION
# ==============================================================

def evaluate_model(actual, predicted):

    predicted = np.maximum(
        np.asarray(predicted).flatten(),
        0
    )

    actual = np.asarray(actual).flatten()

    mae = mean_absolute_error(
        actual, predicted
    )

    mse = mean_squared_error(
        actual, predicted
    )

    rmse = np.sqrt(mse)

    r2 = r2_score(
        actual, predicted
    )

    mape = np.mean(
        np.abs(
            (actual - predicted)
            / np.maximum(np.abs(actual), 1)
        )
    ) * 100

    return {
        "MAE": float(mae),
        "MSE": float(mse),
        "RMSE": float(rmse),
        "R2_Score": float(r2),
        "R2_Percentage": float(r2 * 100),
        "MAPE_Percentage": float(mape)
    }

# ==============================================================
# 11. MACHINE LEARNING MODELS
# ==============================================================

models = {

    "Linear_Regression": LinearRegression(),

    "Random_Forest": RandomForestRegressor(
        n_estimators=200,
        max_depth=20,
        random_state=SEED,
        n_jobs=-1
    ),

    "Gradient_Boosting": GradientBoostingRegressor(
        n_estimators=250,
        learning_rate=0.05,
        max_depth=4,
        random_state=SEED
    ),

    "Extra_Trees": ExtraTreesRegressor(
        n_estimators=200,
        max_depth=25,
        random_state=SEED,
        n_jobs=-1
    )
}

validation_results = {}
test_results = {}

trained_models = {}

print("\nTraining Machine Learning Models...")
print("=" * 70)

for name, model in models.items():

    print(f"\nTraining {name}...")

    model.fit(
        X_train_scaled,
        y_train
    )

    val_prediction = model.predict(
        X_val_scaled
    )

    test_prediction = model.predict(
        X_test_scaled
    )

    validation_results[name] = evaluate_model(
        y_val,
        val_prediction
    )

    test_results[name] = evaluate_model(
        y_test,
        test_prediction
    )

    trained_models[name] = model

    model_path = os.path.join(
        OUTPUT_DIR,
        f"{name.lower()}_model.pkl"
    )

    joblib.dump(model, model_path)

    print(
        "Validation R2:",
        round(
            validation_results[name]["R2_Score"],
            4
        )
    )

    print(
        "Test R2:",
        round(
            test_results[name]["R2_Score"],
            4
        )
    )

    print(
        "Test RMSE:",
        round(
            test_results[name]["RMSE"],
            4
        )
    )

# ==============================================================
# 12. ARTIFICIAL NEURAL NETWORK
# ==============================================================

print("\nTraining Artificial Neural Network...")

# Scale the target for stable ANN training.

target_scaler = StandardScaler()

y_train_scaled = target_scaler.fit_transform(
    y_train.to_numpy().reshape(-1, 1)
)

y_val_scaled = target_scaler.transform(
    y_val.to_numpy().reshape(-1, 1)
)

joblib.dump(
    target_scaler,
    os.path.join(
        OUTPUT_DIR,
        "target_scaler.pkl"
    )
)

ann_model = Sequential([

    Input(shape=(len(FEATURES),)),

    Dense(128, activation="relu"),

    Dropout(0.10),

    Dense(64, activation="relu"),

    Dense(32, activation="relu"),

    Dense(16, activation="relu"),

    Dense(1, activation="linear")
])

ann_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="mse",
    metrics=["mae"]
)

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=20,
    restore_best_weights=True
)

history = ann_model.fit(

    X_train_scaled,
    y_train_scaled,

    validation_data=(
        X_val_scaled,
        y_val_scaled
    ),

    epochs=200,

    batch_size=64,

    callbacks=[early_stopping],

    verbose=1
)

# ==============================================================
# 13. ANN PREDICTIONS AND EVALUATION
# ==============================================================

ann_val_prediction = target_scaler.inverse_transform(
    ann_model.predict(
        X_val_scaled,
        verbose=0
    )
).flatten()

ann_test_prediction = target_scaler.inverse_transform(
    ann_model.predict(
        X_test_scaled,
        verbose=0
    )
).flatten()

validation_results["ANN"] = evaluate_model(
    y_val,
    ann_val_prediction
)

test_results["ANN"] = evaluate_model(
    y_test,
    ann_test_prediction
)

# ==============================================================
# 14. SAVE H5 MODEL
# ==============================================================

H5_PATH = os.path.join(
    OUTPUT_DIR,
    "model.h5"
)

ann_model.save(H5_PATH)

print("\nANN Model Saved:", H5_PATH)

# Also save training history for future visualization.

history_df = pd.DataFrame(history.history)

history_df.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "training_history.csv"
    ),
    index=False
)

# ==============================================================
# 15. SELECT BEST MACHINE LEARNING MODEL
# ==============================================================

# Select using validation performance, not test performance.

best_ml_name = max(
    models.keys(),
    key=lambda name:
        validation_results[name]["R2_Score"]
)

best_ml_model = trained_models[best_ml_name]

best_overall_name = max(
    validation_results,
    key=lambda name:
        validation_results[name]["R2_Score"]
)

print("\nBest ML Model:", best_ml_name)

print("Best Overall Model:", best_overall_name)

# ==============================================================
# 16. SAVE PKL MODEL
# ==============================================================

PKL_PATH = os.path.join(
    OUTPUT_DIR,
    "model.pkl"
)

joblib.dump(
    best_ml_model,
    PKL_PATH
)

# Save complete preprocessing and model bundle.

model_bundle = {

    "model_name": best_ml_name,

    "model": best_ml_model,

    "scaler": scaler,

    "features": FEATURES,

    "target": TARGET,

    "data_type": "synthetic_simulation"
}

joblib.dump(
    model_bundle,
    os.path.join(
        OUTPUT_DIR,
        "model_bundle.pkl"
    )
)

print("\nBest ML Model Saved:", PKL_PATH)

# ==============================================================
# 17. SAVE YAML CONFIGURATION
# ==============================================================

configuration = {

    "project": {
        "name": (
            "AI-Based Municipal Waste Management "
            "and Conservancy Workforce Optimization"
        ),
        "version": "1.0",
        "task": "regression",
        "random_seed": SEED,
        "data_type": "synthetic_simulation"
    },

    "dataset": {
        "original_dataset": DATASET_PATH,
        "original_records": int(len(df)),
        "synthetic_records": N_SAMPLES,
        "synthetic_dataset": (
            "synthetic_municipal_dataset.csv"
        ),
        "target": TARGET,
        "features": FEATURES
    },

    "baseline": {
        "conservancy_workers": BASE_WORKERS,
        "disposal_sites": BASE_SITES,
        "other_workers": BASE_OTHER_WORKERS,
        "total_workforce": TOTAL_WORKFORCE,
        "workers_per_site": float(WORKERS_PER_SITE)
    },

    "simulation": {
        "nominal_capacity_tonnes_per_worker_hour": 0.18,
        "minimum_workers_per_site": 2,
        "noise_standard_deviation": 0.03,
        "waste_per_site_range": [3.0, 12.0],
        "working_hours_range": [6, 10],
        "productivity_range": [0.6, 1.4],
        "equipment_efficiency_range": [0.55, 0.98],
        "absenteeism_range": [0.02, 0.20],
        "transport_distance_range_km": [2, 30],
        "seasonal_demand_factor_range": [0.8, 1.5],
        "daily_wage_range": [400, 1000]
    },

    "training": {
        "training_samples": int(len(X_train)),
        "validation_samples": int(len(X_val)),
        "testing_samples": int(len(X_test)),
        "feature_scaling": "StandardScaler",
        "target_scaling_for_ann": "StandardScaler",
        "epochs": 200,
        "batch_size": 64,
        "early_stopping_patience": 20
    },

    "models": {
        "best_machine_learning_model": best_ml_name,
        "best_overall_model": best_overall_name,
        "ann_model": "model.h5",
        "machine_learning_model": "model.pkl",
        "model_bundle": "model_bundle.pkl",
        "scaler": "scaler.pkl",
        "ann_target_scaler": "target_scaler.pkl"
    },

    "limitations": [
        "Original dataset contains one aggregate record.",
        "Operational variables are synthetically generated.",
        "Prediction targets follow an assumed simulation formula.",
        "Results are not validated against real municipal operations."
    ]
}

YAML_PATH = os.path.join(
    OUTPUT_DIR,
    "config.yaml"
)

with open(
    YAML_PATH,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        configuration,
        file,
        sort_keys=False,
        allow_unicode=True
    )

print("\nYAML Configuration Saved:", YAML_PATH)

# ==============================================================
# 18. SAVE JSON PERFORMANCE METRICS
# ==============================================================

metrics = {

    "project": (
        "AI-Based Municipal Waste Management "
        "and Conservancy Workforce Optimization"
    ),

    "evaluation_type": (
        "Held-out synthetic scenario evaluation"
    ),

    "original_dataset_records": int(len(df)),

    "synthetic_dataset_records": N_SAMPLES,

    "baseline_statistics": {
        "conservancy_workers": BASE_WORKERS,
        "disposal_sites": BASE_SITES,
        "other_workers": BASE_OTHER_WORKERS,
        "total_workforce": TOTAL_WORKFORCE,
        "workers_per_site": float(WORKERS_PER_SITE)
    },

    "best_machine_learning_model": best_ml_name,

    "best_overall_model": best_overall_name,

    "validation_results": validation_results,

    "test_results": test_results,

    "best_model_test_performance": (
        test_results[best_overall_name]
    ),

    "real_world_validated": False
}

JSON_PATH = os.path.join(
    OUTPUT_DIR,
    "metrics.json"
)

with open(
    JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        metrics,
        file,
        indent=4,
        allow_nan=False
    )

print("\nJSON Metrics Saved:", JSON_PATH)

# ==============================================================
# 19. SAVE MODEL ARCHITECTURE JSON
# ==============================================================

with open(
    os.path.join(
        OUTPUT_DIR,
        "model_architecture.json"
    ),
    "w",
    encoding="utf-8"
) as file:

    file.write(
        ann_model.to_json()
    )

# ==============================================================
# 20. MODEL COMPARISON
# ==============================================================

comparison_df = pd.DataFrame(
    test_results
).T

comparison_df.index.name = "Model"

comparison_df = comparison_df.sort_values(
    by="R2_Score",
    ascending=False
)

comparison_df.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "model_comparison.csv"
    )
)

print("\n" + "=" * 70)
print("MODEL PERFORMANCE COMPARISON")
print("=" * 70)

print(comparison_df.round(4))

# ==============================================================
# 21. GENERATE TEST PREDICTIONS
# ==============================================================

prediction_df = synthetic_df.loc[
    X_test.index
].copy()

prediction_df["actual_required_workers"] = (
    y_test.to_numpy()
)

for name, model in trained_models.items():

    prediction_df[
        f"{name}_prediction"
    ] = np.maximum(
        np.ceil(
            model.predict(X_test_scaled)
        ),
        2 * prediction_df["disposal_sites"].to_numpy()
    ).astype(int)

prediction_df["ANN_prediction"] = np.maximum(
    np.ceil(ann_test_prediction),
    2 * prediction_df["disposal_sites"].to_numpy()
).astype(int)

prediction_df["predicted_required_workers"] = (
    prediction_df[
        f"{best_overall_name}_prediction"
    ]
)

prediction_df["predicted_workforce_shortage"] = (
    np.maximum(
        prediction_df["predicted_required_workers"]
        - prediction_df["available_workers"],
        0
    )
)

prediction_df["predicted_workforce_surplus"] = (
    np.maximum(
        prediction_df["available_workers"]
        - prediction_df["predicted_required_workers"],
        0
    )
)

prediction_df["predicted_daily_cost"] = (
    prediction_df["predicted_required_workers"]
    * prediction_df["daily_wage"]
)

prediction_df["allocation_recommendation"] = np.select(

    [
        prediction_df["predicted_workforce_shortage"] > 0,
        prediction_df["predicted_workforce_surplus"] > 0
    ],

    [
        "Additional Workforce Required",
        "Workforce Redistribution Recommended"
    ],

    default="Workforce Balanced"
)

prediction_df.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "prediction.csv"
    ),
    index=False
)

print("\nPrediction Preview:")

print(
    prediction_df[[
        "disposal_sites",
        "available_workers",
        "actual_required_workers",
        "predicted_required_workers",
        "predicted_workforce_shortage",
        "allocation_recommendation"
    ]].head(10)
)

# ==============================================================
# 22. PREDICT NEW MUNICIPAL SCENARIO
# ==============================================================

new_scenario = pd.DataFrame([{

    "disposal_sites": BASE_SITES,

    "other_workers": BASE_OTHER_WORKERS,

    "daily_waste_tonnes": 650,

    "population_density": 12000,

    "working_hours": 8,

    "worker_productivity": 1.0,

    "equipment_efficiency": 0.85,

    "absenteeism_rate": 0.08,

    "transport_distance_km": 10,

    "seasonal_demand_factor": 1.1
}])

new_scenario_scaled = scaler.transform(
    new_scenario[FEATURES]
)

if best_overall_name == "ANN":

    predicted_workers = (
        target_scaler.inverse_transform(
            ann_model.predict(
                new_scenario_scaled,
                verbose=0
            )
        )[0][0]
    )

else:

    predicted_workers = best_ml_model.predict(
        new_scenario_scaled
    )[0]

predicted_workers = max(
    int(np.ceil(predicted_workers)),
    BASE_SITES * 2
)

shortage = max(
    predicted_workers - BASE_WORKERS,
    0
)

surplus = max(
    BASE_WORKERS - predicted_workers,
    0
)

print("\n" + "=" * 70)
print("NEW SIMULATED MUNICIPAL SCENARIO")
print("=" * 70)

print("Disposal Sites:", BASE_SITES)

print("Available Workers:", BASE_WORKERS)

print("Predicted Required Workers:", predicted_workers)

print("Workforce Shortage:", shortage)

print("Workforce Surplus:", surplus)

if shortage > 0:

    print(
        f"Recommendation: Allocate "
        f"{shortage} additional workers."
    )

elif surplus > 0:

    print(
        f"Recommendation: Consider redistributing "
        f"{surplus} workers."
    )

else:

    print("Recommendation: Maintain current workforce.")

# ==============================================================
# 23. DISPLAY GENERATED FILES
# ==============================================================

print("\n" + "=" * 70)
print("PROJECT EXECUTED SUCCESSFULLY")
print("=" * 70)

print("\nOutput Directory:")
print(OUTPUT_DIR)

print("\nGenerated and Available Files:")

for filename in sorted(os.listdir(OUTPUT_DIR)):

    filepath = os.path.join(
        OUTPUT_DIR,
        filename
    )

    if os.path.isfile(filepath):

        size_kb = os.path.getsize(
            filepath
        ) / 1024

        print(
            f"{filename:<45} "
            f"{size_kb:>10.2f} KB"
        )

print("\nIMPORTANT:")

print(
    "All model performance metrics are based on "
    "synthetic operational scenarios."
)

print(
    "Real-world validation requires additional "
    "municipal workforce and waste management data."
)

print("\nAll model and configuration files saved successfully.")
```

```python
import os
import time
import json
import yaml
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense

# ==============================================================
# 1. CONFIGURATION
# ==============================================================

START_TIME = time.time()

SEED = 42

np.random.seed(SEED)
tf.random.set_seed(SEED)

rng = np.random.default_rng(SEED)

OUTPUT_DIR = r"C:\Users\sagni\Downloads\AI-Based Municipal Waste Management and Conservancy Workforce"

ORIGINAL_DATASET = os.path.join(
    OUTPUT_DIR,
    "number_of conservancy_workers_engaged_in_conservancy_garbage_and_debris_disposal_and_other_particulars_2017_18.csv"
)

SYNTHETIC_DATASET = os.path.join(
    OUTPUT_DIR,
    "synthetic_municipal_dataset.csv"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

sns.set_theme(style="whitegrid")

plt.rcParams["figure.dpi"] = 120
plt.rcParams["savefig.dpi"] = 300
plt.rcParams["font.size"] = 11

# AIS hyperparameters

POPULATION_SIZE = 40
GENERATIONS = 150
SELECTED_ANTIBODIES = 10
CLONES_PER_ANTIBODY = 5
RECEPTOR_EDITING_RATE = 0.15

INITIAL_MUTATION_RATE = 0.15
MIN_MUTATION_RATE = 0.005

HIDDEN_LAYER_1 = 16
HIDDEN_LAYER_2 = 8

WEIGHT_LIMIT = 5.0
REGULARIZATION = 0.0001

print("=" * 75)
print("AIS-BASED MUNICIPAL WASTE MANAGEMENT")
print("CLONAL SELECTION ALGORITHM")
print("=" * 75)

# ==============================================================
# 2. LOAD DATASET
# ==============================================================

if not os.path.exists(SYNTHETIC_DATASET):

    raise FileNotFoundError(
        "Synthetic dataset not found. "
        "Run the previous baseline training code first."
    )

df = pd.read_csv(SYNTHETIC_DATASET)

print("\nDataset Shape:", df.shape)

print("\nDataset Preview:")
print(df.head())

# Save AIS-specific dataset copy

df.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "ais_dataset.csv"
    ),
    index=False
)

# ==============================================================
# 3. FEATURE AND TARGET SELECTION
# ==============================================================

FEATURES = [

    "disposal_sites",

    "other_workers",

    "daily_waste_tonnes",

    "population_density",

    "working_hours",

    "worker_productivity",

    "equipment_efficiency",

    "absenteeism_rate",

    "transport_distance_km",

    "seasonal_demand_factor"
]

TARGET = "required_workers"

required_columns = FEATURES + [

    TARGET,

    "available_workers",

    "daily_wage"
]

missing_columns = [

    col for col in required_columns

    if col not in df.columns
]

if missing_columns:

    raise ValueError(
        f"Missing columns: {missing_columns}"
    )

if df[required_columns].isna().any().any():

    raise ValueError(
        "Missing values detected in required columns."
    )

X = df[FEATURES]
y = df[TARGET]

print("\nFeatures:", FEATURES)
print("\nTarget:", TARGET)

# ==============================================================
# 4. TRAIN, VALIDATION AND TEST SPLIT
# ==============================================================

# Same split as the baseline implementation.

X_train_full, X_test, y_train_full, y_test = (
    train_test_split(

        X,

        y,

        test_size=0.20,

        random_state=SEED
    )
)

X_train, X_val, y_train, y_val = (
    train_test_split(

        X_train_full,

        y_train_full,

        test_size=0.20,

        random_state=SEED
    )
)

print("\nTraining Samples:", len(X_train))
print("Validation Samples:", len(X_val))
print("Testing Samples:", len(X_test))

# ==============================================================
# 5. FEATURE AND TARGET SCALING
# ==============================================================

feature_scaler = StandardScaler()

X_train_scaled = feature_scaler.fit_transform(
    X_train
)

X_val_scaled = feature_scaler.transform(
    X_val
)

X_test_scaled = feature_scaler.transform(
    X_test
)

target_scaler = StandardScaler()

y_train_scaled = target_scaler.fit_transform(

    y_train.to_numpy().reshape(-1, 1)

).flatten()

y_val_scaled = target_scaler.transform(

    y_val.to_numpy().reshape(-1, 1)

).flatten()

# Use float32 for faster AIS computation.

X_train_scaled = X_train_scaled.astype(
    np.float32
)

X_val_scaled = X_val_scaled.astype(
    np.float32
)

X_test_scaled = X_test_scaled.astype(
    np.float32
)

y_train_scaled = y_train_scaled.astype(
    np.float32
)

y_val_scaled = y_val_scaled.astype(
    np.float32
)

joblib.dump(

    feature_scaler,

    os.path.join(
        OUTPUT_DIR,
        "ais_scaler.pkl"
    )
)

joblib.dump(

    target_scaler,

    os.path.join(
        OUTPUT_DIR,
        "ais_target_scaler.pkl"
    )
)

# ==============================================================
# 6. NEURAL NETWORK ARCHITECTURE
# ==============================================================

INPUT_SIZE = len(FEATURES)

WEIGHT_SHAPES = [

    (INPUT_SIZE, HIDDEN_LAYER_1),

    (HIDDEN_LAYER_1,),

    (HIDDEN_LAYER_1, HIDDEN_LAYER_2),

    (HIDDEN_LAYER_2,),

    (HIDDEN_LAYER_2, 1),

    (1,)
]

TOTAL_PARAMETERS = sum(

    int(np.prod(shape))

    for shape in WEIGHT_SHAPES
)

print("\nNeural Network Architecture:")

print(
    INPUT_SIZE,
    "->",
    HIDDEN_LAYER_1,
    "->",
    HIDDEN_LAYER_2,
    "-> 1"
)

print("Total AIS Parameters:", TOTAL_PARAMETERS)

# ==============================================================
# 7. ANTIBODY REPRESENTATION
# ==============================================================

# Each antibody represents all neural network
# weights and biases.

def decode_antibody(antibody):

    weights = []

    start = 0

    for shape in WEIGHT_SHAPES:

        size = int(np.prod(shape))

        parameter = antibody[
            start:start + size
        ].reshape(shape)

        weights.append(parameter)

        start += size

    return weights

# ==============================================================
# 8. NEURAL NETWORK FORWARD PROPAGATION
# ==============================================================

def relu(x):

    return np.maximum(x, 0)


def forward_pass(X_data, antibody):

    W1, b1, W2, b2, W3, b3 = (
        decode_antibody(antibody)
    )

    hidden1 = relu(
        X_data @ W1 + b1
    )

    hidden2 = relu(
        hidden1 @ W2 + b2
    )

    output = (
        hidden2 @ W3 + b3
    )

    return output.flatten()

# ==============================================================
# 9. FITNESS AND AFFINITY FUNCTIONS
# ==============================================================

def fitness_function(

    antibody,

    X_data,

    y_data,

    regularize=True
):

    predictions = forward_pass(
        X_data,
        antibody
    )

    mse = np.mean(
        (y_data - predictions) ** 2
    )

    penalty = (

        REGULARIZATION
        * np.mean(antibody ** 2)

        if regularize else 0.0
    )

    return float(mse + penalty)


def affinity_function(fitness):

    return 1.0 / (
        1.0 + fitness
    )

# ==============================================================
# 10. INITIALIZE ANTIBODY POPULATION
# ==============================================================

def create_antibody():

    parameters = []

    for shape in WEIGHT_SHAPES:

        if len(shape) == 2:

            std = np.sqrt(
                2.0 / shape[0]
            )

            values = rng.normal(
                0,
                std,
                size=shape
            )

        else:

            values = np.zeros(
                shape
            )

        parameters.append(
            values.flatten()
        )

    return np.concatenate(
        parameters
    ).astype(np.float32)


population = np.array([

    create_antibody()

    for _ in range(POPULATION_SIZE)

], dtype=np.float32)

print("\nInitial Population Shape:")

print(population.shape)

# ==============================================================
# 11. CLONAL SELECTION AND HYPERMUTATION
# ==============================================================

def hypermutation(

    antibody,

    affinity,

    generation
):

    progress = (
        generation / GENERATIONS
    )

    base_mutation = max(

        MIN_MUTATION_RATE,

        INITIAL_MUTATION_RATE
        * (1.0 - progress)
    )

    # High-affinity antibodies receive
    # smaller mutations.

    mutation_strength = max(

        MIN_MUTATION_RATE,

        base_mutation
        * np.exp(-2.0 * affinity)
    )

    mutation_mask = (

        rng.random(TOTAL_PARAMETERS)
        < 0.35
    )

    noise = rng.normal(

        0,

        mutation_strength,

        TOTAL_PARAMETERS

    ).astype(np.float32)

    clone = antibody.copy()

    clone[mutation_mask] += (
        noise[mutation_mask]
    )

    clone = np.clip(

        clone,

        -WEIGHT_LIMIT,

        WEIGHT_LIMIT
    )

    return clone.astype(
        np.float32
    )

# ==============================================================
# 12. AIS OPTIMIZATION
# ==============================================================

best_antibody = None

best_validation_loss = float("inf")

best_generation = 0

convergence_history = []

print("\nStarting AIS Optimization...")
print("=" * 75)

for generation in range(

    1,

    GENERATIONS + 1
):

    # ----------------------------------------------------------
    # A. Evaluate antibody population
    # ----------------------------------------------------------

    fitness_values = np.array([

        fitness_function(

            antibody,

            X_train_scaled,

            y_train_scaled
        )

        for antibody in population
    ])

    ranked_indices = np.argsort(
        fitness_values
    )

    population = population[
        ranked_indices
    ]

    fitness_values = fitness_values[
        ranked_indices
    ]

    # ----------------------------------------------------------
    # B. Select high-affinity antibodies
    # ----------------------------------------------------------

    selected_population = population[
        :SELECTED_ANTIBODIES
    ]

    selected_fitness = fitness_values[
        :SELECTED_ANTIBODIES
    ]

    # ----------------------------------------------------------
    # C. Clone selected antibodies
    # ----------------------------------------------------------

    clones = []

    for rank, antibody in enumerate(
        selected_population
    ):

        affinity = affinity_function(
            selected_fitness[rank]
        )

        # Better-ranked antibodies generate
        # additional clones.

        clone_count = (

            CLONES_PER_ANTIBODY

            + SELECTED_ANTIBODIES

            - rank
        )

        for _ in range(clone_count):

            clone = hypermutation(

                antibody,

                affinity,

                generation
            )

            clones.append(clone)

    clones = np.asarray(

        clones,

        dtype=np.float32
    )

    # ----------------------------------------------------------
    # D. Evaluate mutated clones
    # ----------------------------------------------------------

    clone_fitness = np.array([

        fitness_function(

            clone,

            X_train_scaled,

            y_train_scaled
        )

        for clone in clones
    ])

    # ----------------------------------------------------------
    # E. Combine parent antibodies and clones
    # ----------------------------------------------------------

    combined_population = np.vstack([

        population,

        clones
    ])

    combined_fitness = np.concatenate([

        fitness_values,

        clone_fitness
    ])

    combined_rank = np.argsort(
        combined_fitness
    )

    population = combined_population[

        combined_rank[:POPULATION_SIZE]

    ].copy()

    # ----------------------------------------------------------
    # F. Immune memory using validation data
    # ----------------------------------------------------------

    generation_best = population[0]

    generation_training_loss = fitness_function(

        generation_best,

        X_train_scaled,

        y_train_scaled,

        regularize=False
    )

    generation_validation_loss = fitness_function(

        generation_best,

        X_val_scaled,

        y_val_scaled,

        regularize=False
    )

    if generation_validation_loss < best_validation_loss:

        best_validation_loss = (
            generation_validation_loss
        )

        best_antibody = (
            generation_best.copy()
        )

        best_generation = generation

    # ----------------------------------------------------------
    # G. Receptor editing
    # ----------------------------------------------------------

    replacement_count = max(

        1,

        int(
            POPULATION_SIZE
            * RECEPTOR_EDITING_RATE
        )
    )

    replacement_count = min(

        replacement_count,

        POPULATION_SIZE - 1
    )

    for i in range(replacement_count):

        population[-(i + 1)] = (
            create_antibody()
        )

    # ----------------------------------------------------------
    # H. Record convergence
    # ----------------------------------------------------------

    convergence_history.append({

        "Generation": generation,

        "Training_Loss": float(
            generation_training_loss
        ),

        "Validation_Loss": float(
            generation_validation_loss
        ),

        "Best_Validation_Loss": float(
            best_validation_loss
        ),

        "Best_Affinity": float(
            affinity_function(
                generation_training_loss
            )
        ),

        "Population_Diversity": float(
            np.mean(
                np.std(population, axis=0)
            )
        ),

        "Number_of_Clones": len(clones)
    })

    if generation == 1 or generation % 10 == 0:

        print(

            f"Generation {generation:03d}/{GENERATIONS} | "

            f"Training Loss: {generation_training_loss:.6f} | "

            f"Validation Loss: {generation_validation_loss:.6f} | "

            f"Best Loss: {best_validation_loss:.6f}"
        )

print("\nAIS Optimization Completed.")

print("Best Generation:", best_generation)

print(
    "Best Validation Loss:",
    best_validation_loss
)

# ==============================================================
# 13. SAVE AIS CONVERGENCE HISTORY
# ==============================================================

history_df = pd.DataFrame(
    convergence_history
)

history_df.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "ais_convergence_history.csv"
    ),

    index=False
)

np.save(

    os.path.join(
        OUTPUT_DIR,
        "ais_best_antibody.npy"
    ),

    best_antibody
)

# ==============================================================
# 14. CREATE KERAS MODEL USING AIS WEIGHTS
# ==============================================================

ais_model = Sequential([

    Input(
        shape=(INPUT_SIZE,)
    ),

    Dense(
        HIDDEN_LAYER_1,
        activation="relu"
    ),

    Dense(
        HIDDEN_LAYER_2,
        activation="relu"
    ),

    Dense(
        1,
        activation="linear"
    )
])

# No gradient-based training is performed.
# All weights are obtained from AIS.

ais_model.set_weights(
    decode_antibody(best_antibody)
)

ais_model.compile(

    optimizer="adam",

    loss="mse",

    metrics=["mae"]
)

ais_model.save(

    os.path.join(
        OUTPUT_DIR,
        "ais_model.h5"
    )
)

print("\nAIS H5 Model Saved.")

# ==============================================================
# 15. SAVE PKL MODEL BUNDLE
# ==============================================================

ais_model_bundle = {

    "algorithm": "Artificial Immune System",

    "optimization_method": "Clonal Selection",

    "best_antibody": best_antibody,

    "weight_shapes": WEIGHT_SHAPES,

    "weights": decode_antibody(
        best_antibody
    ),

    "feature_scaler": feature_scaler,

    "target_scaler": target_scaler,

    "features": FEATURES,

    "target": TARGET,

    "architecture": [

        INPUT_SIZE,

        HIDDEN_LAYER_1,

        HIDDEN_LAYER_2,

        1
    ],

    "best_generation": best_generation,

    "data_type": "synthetic_simulation"
}

joblib.dump(

    ais_model_bundle,

    os.path.join(
        OUTPUT_DIR,
        "ais_model.pkl"
    )
)

print("AIS PKL Model Saved.")

# ==============================================================
# 16. GENERATE WORKFORCE PREDICTIONS
# ==============================================================

def predict_workforce(

    X_scaled,

    X_original
):

    scaled_predictions = forward_pass(

        X_scaled,

        best_antibody
    )

    predictions = target_scaler.inverse_transform(

        scaled_predictions.reshape(-1, 1)

    ).flatten()

    minimum_workers = (

        2 * X_original[
            "disposal_sites"
        ].to_numpy()
    )

    predictions = np.maximum(

        np.ceil(predictions),

        minimum_workers
    )

    return predictions.astype(int)


train_predictions = predict_workforce(

    X_train_scaled,

    X_train
)

val_predictions = predict_workforce(

    X_val_scaled,

    X_val
)

test_predictions = predict_workforce(

    X_test_scaled,

    X_test
)

# ==============================================================
# 17. PERFORMANCE EVALUATION
# ==============================================================

def evaluate_model(actual, predicted):

    actual = np.asarray(
        actual
    ).flatten()

    predicted = np.asarray(
        predicted
    ).flatten()

    mae = mean_absolute_error(
        actual,
        predicted
    )

    mse = mean_squared_error(
        actual,
        predicted
    )

    rmse = np.sqrt(mse)

    r2 = r2_score(
        actual,
        predicted
    )

    mape = np.mean(

        np.abs(

            (actual - predicted)

            / np.maximum(
                np.abs(actual),
                1
            )
        )

    ) * 100

    return {

        "MAE": float(mae),

        "MSE": float(mse),

        "RMSE": float(rmse),

        "R2_Score": float(r2),

        "R2_Percentage": float(
            r2 * 100
        ),

        "MAPE_Percentage": float(mape)
    }


train_metrics = evaluate_model(

    y_train,

    train_predictions
)

validation_metrics = evaluate_model(

    y_val,

    val_predictions
)

test_metrics = evaluate_model(

    y_test,

    test_predictions
)

print("\n" + "=" * 75)
print("AIS TEST PERFORMANCE")
print("=" * 75)

for metric, value in test_metrics.items():

    print(
        f"{metric:<25}: {value:.6f}"
    )

# ==============================================================
# 18. SAVE RESULT CSV
# ==============================================================

result_df = pd.DataFrame([

    {

        "Model": "AIS-ANN",

        "Dataset": "Training",

        **train_metrics
    },

    {

        "Model": "AIS-ANN",

        "Dataset": "Validation",

        **validation_metrics
    },

    {

        "Model": "AIS-ANN",

        "Dataset": "Testing",

        **test_metrics
    }
])

result_df["Evaluation_Type"] = (
    "Synthetic Simulation"
)

result_df.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "ais_result.csv"
    ),

    index=False
)

print("\nAIS Result DataFrame:")

print(
    result_df.round(4).to_string(
        index=False
    )
)

# ==============================================================
# 19. CREATE PREDICTION DATAFRAME
# ==============================================================

prediction_df = df.loc[
    X_test.index
].copy()

prediction_df.insert(

    0,

    "Scenario_ID",

    X_test.index.to_numpy() + 1
)

prediction_df["Actual_Required_Workers"] = (
    y_test.to_numpy()
)

prediction_df["AIS_Predicted_Required_Workers"] = (
    test_predictions
)

prediction_df["Absolute_Error"] = np.abs(

    prediction_df[
        "Actual_Required_Workers"
    ]

    - prediction_df[
        "AIS_Predicted_Required_Workers"
    ]
)

prediction_df["Percentage_Error"] = (

    prediction_df["Absolute_Error"]

    / np.maximum(

        prediction_df[
            "Actual_Required_Workers"
        ],

        1
    )

) * 100

# ==============================================================
# 20. WORKFORCE REQUIREMENT ANALYSIS
# ==============================================================

prediction_df["AIS_Workforce_Shortage"] = (
    np.maximum(

        prediction_df[
            "AIS_Predicted_Required_Workers"
        ]

        - prediction_df[
            "available_workers"
        ],

        0
    )
)

prediction_df["AIS_Workforce_Surplus"] = (
    np.maximum(

        prediction_df[
            "available_workers"
        ]

        - prediction_df[
            "AIS_Predicted_Required_Workers"
        ],

        0
    )
)

prediction_df["AIS_Workforce_Utilization"] = (

    np.minimum(

        prediction_df[
            "AIS_Predicted_Required_Workers"
        ]

        / np.maximum(

            prediction_df[
                "available_workers"
            ],

            1
        ),

        1.0
    )

    * 100
)

prediction_df["AIS_Estimated_Daily_Cost"] = (

    prediction_df[
        "AIS_Predicted_Required_Workers"
    ]

    * prediction_df[
        "daily_wage"
    ]
)

prediction_df["AIS_Additional_Workforce_Cost"] = (

    prediction_df[
        "AIS_Workforce_Shortage"
    ]

    * prediction_df[
        "daily_wage"
    ]
)

prediction_df["AIS_Workforce_Status"] = np.select(

    [

        prediction_df[
            "AIS_Workforce_Shortage"
        ] > 0,

        prediction_df[
            "AIS_Workforce_Surplus"
        ] > 0
    ],

    [

        "Workforce Shortage",

        "Workforce Surplus"
    ],

    default="Balanced Workforce"
)

prediction_df["AIS_Recommendation"] = np.select(

    [

        prediction_df[
            "AIS_Workforce_Shortage"
        ] > 0,

        prediction_df[
            "AIS_Workforce_Surplus"
        ] > 0
    ],

    [

        "Allocate Additional Workers",

        "Consider Workforce Redistribution"
    ],

    default="Maintain Current Workforce"
)

prediction_df["Data_Type"] = (
    "Synthetic Simulation"
)

prediction_df.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "ais_prediction.csv"
    ),

    index=False
)

print("\nAIS Prediction Preview:")

print(

    prediction_df[[

        "Scenario_ID",

        "available_workers",

        "Actual_Required_Workers",

        "AIS_Predicted_Required_Workers",

        "Absolute_Error",

        "AIS_Workforce_Status"

    ]].head(10).to_string(index=False)
)

# ==============================================================
# 21. GRAPH SAVE AND DISPLAY FUNCTION
# ==============================================================

def save_show(filename):

    plt.tight_layout()

    plt.savefig(

        os.path.join(
            OUTPUT_DIR,
            filename
        ),

        dpi=300,

        bbox_inches="tight",

        facecolor="white"
    )

    plt.show()

    plt.close()

    print("Saved:", filename)

# ==============================================================
# 22. AIS ACCURACY GRAPH
# ==============================================================

plt.figure(figsize=(10, 6))

labels = [

    "Training",

    "Validation",

    "Testing"
]

scores = [

    train_metrics["R2_Percentage"],

    validation_metrics["R2_Percentage"],

    test_metrics["R2_Percentage"]
]

bars = plt.bar(

    labels,

    scores,

    color=[

        "#3498db",

        "#e67e22",

        "#2ecc71"
    ],

    edgecolor="black",

    width=0.55
)

for bar, value in zip(bars, scores):

    plt.annotate(

        f"{value:.2f}%",

        (

            bar.get_x()
            + bar.get_width() / 2,

            value
        ),

        ha="center",

        va="bottom" if value >= 0 else "top",

        xytext=(0, 5 if value >= 0 else -5),

        textcoords="offset points",

        fontweight="bold"
    )

plt.ylim(

    min(0, min(scores) - 10),

    max(110, max(scores) + 10)
)

plt.axhline(
    0,
    color="black"
)

plt.title(

    "AIS-Based Municipal Workforce Prediction - R² Performance",

    fontsize=15,

    fontweight="bold"
)

plt.xlabel("Dataset Split")

plt.ylabel("R² Score (%)")

save_show("ais_accuracy_graph.png")

# ==============================================================
# 23. AIS CORRELATION HEATMAP
# ==============================================================

correlation_matrix = df[

    FEATURES + [TARGET]

].corr()

plt.figure(figsize=(15, 11))

sns.heatmap(

    correlation_matrix,

    annot=True,

    fmt=".2f",

    cmap="coolwarm",

    center=0,

    vmin=-1,

    vmax=1,

    linewidths=0.5
)

plt.title(

    "AIS Municipal Workforce - Feature Correlation Heatmap",

    fontsize=16,

    fontweight="bold"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.yticks(rotation=0)

save_show("ais_heatmap.png")

correlation_matrix.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "ais_correlation_matrix.csv"
    )
)

# ==============================================================
# 24. AIS VS BASELINE MODEL COMPARISON
# ==============================================================

BASELINE_RESULT_PATH = os.path.join(

    OUTPUT_DIR,

    "result.csv"
)

comparison_columns = [

    "Model",

    "MAE",

    "RMSE",

    "R2_Score",

    "R2_Percentage",

    "MAPE_Percentage"
]

ais_comparison_row = pd.DataFrame([{

    "Model": "AIS-ANN",

    **test_metrics
}])[comparison_columns]

if os.path.exists(BASELINE_RESULT_PATH):

    baseline_df = pd.read_csv(
        BASELINE_RESULT_PATH
    )

    if set(comparison_columns).issubset(
        baseline_df.columns
    ):

        comparison_df = pd.concat(

            [

                baseline_df[
                    comparison_columns
                ],

                ais_comparison_row
            ],

            ignore_index=True
        )

    else:

        print(
            "Baseline result columns do not match."
        )

        comparison_df = (
            ais_comparison_row.copy()
        )

else:

    print(
        "Baseline result.csv not found. "
        "Only AIS results will be plotted."
    )

    comparison_df = (
        ais_comparison_row.copy()
    )

comparison_df = comparison_df.sort_values(

    "R2_Score",

    ascending=False
)

comparison_df.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "ais_comparison.csv"
    ),

    index=False
)

print("\nAIS Model Comparison:")

print(
    comparison_df.round(4).to_string(
        index=False
    )
)

fig, axes = plt.subplots(

    2,

    2,

    figsize=(16, 11)
)

comparison_metrics = [

    ("R2_Percentage", "R² Score (%)"),

    ("MAE", "Mean Absolute Error"),

    ("RMSE", "Root Mean Squared Error"),

    ("MAPE_Percentage", "MAPE (%)")
]

colors = [

    "#e74c3c"

    if name == "AIS-ANN"

    else "#3498db"

    for name in comparison_df["Model"]
]

for ax, (metric, title) in zip(

    axes.flatten(),

    comparison_metrics
):

    bars = ax.bar(

        comparison_df["Model"],

        comparison_df[metric],

        color=colors,

        edgecolor="black"
    )

    ax.set_title(

        title,

        fontweight="bold"
    )

    ax.tick_params(

        axis="x",

        rotation=20
    )

    ax.margins(y=0.15)

    for bar, value in zip(

        bars,

        comparison_df[metric]
    ):

        ax.annotate(

            f"{value:.2f}",

            (

                bar.get_x()
                + bar.get_width() / 2,

                value
            ),

            ha="center",

            va="bottom" if value >= 0 else "top",

            fontsize=9
        )

fig.suptitle(

    "AIS vs Conventional Machine Learning Models",

    fontsize=18,

    fontweight="bold"
)

save_show("ais_comparison_graph.png")

# ==============================================================
# 25. AIS RESULT GRAPH
# ==============================================================

fig, axes = plt.subplots(

    2,

    2,

    figsize=(13, 10)
)

metrics_to_plot = [

    ("MAE", "Mean Absolute Error"),

    ("RMSE", "Root Mean Squared Error"),

    ("R2_Score", "R² Score"),

    ("MAPE_Percentage", "MAPE (%)")
]

for ax, (metric, title) in zip(

    axes.flatten(),

    metrics_to_plot
):

    bars = ax.bar(

        result_df["Dataset"],

        result_df[metric],

        color=[

            "#3498db",

            "#e67e22",

            "#2ecc71"
        ],

        edgecolor="black"
    )

    ax.set_title(

        title,

        fontweight="bold"
    )

    ax.margins(y=0.15)

    for bar, value in zip(

        bars,

        result_df[metric]
    ):

        ax.annotate(

            f"{value:.3f}",

            (

                bar.get_x()
                + bar.get_width() / 2,

                value
            ),

            ha="center",

            va="bottom" if value >= 0 else "top"
        )

fig.suptitle(

    "AIS Training, Validation and Testing Results",

    fontsize=17,

    fontweight="bold"
)

save_show("ais_result_graph.png")

# ==============================================================
# 26. AIS PREDICTION GRAPH
# ==============================================================

N_DISPLAY = min(

    100,

    len(prediction_df)
)

sample = prediction_df.head(
    N_DISPLAY
)

positions = np.arange(

    1,

    N_DISPLAY + 1
)

plt.figure(figsize=(16, 7))

plt.plot(

    positions,

    sample[
        "Actual_Required_Workers"
    ],

    label="Simulated Actual Workforce",

    color="#3498db",

    linewidth=2,

    marker="o",

    markersize=3
)

plt.plot(

    positions,

    sample[
        "AIS_Predicted_Required_Workers"
    ],

    label="AIS Predicted Workforce",

    color="#e74c3c",

    linestyle="--",

    linewidth=2,

    marker="x",

    markersize=3
)

plt.title(

    "AIS - Actual vs Predicted Municipal Workforce",

    fontsize=16,

    fontweight="bold"
)

plt.xlabel("Test Scenario")

plt.ylabel("Required Workers")

plt.legend()

save_show("ais_prediction_graph.png")

# ==============================================================
# 27. AIS CONVERGENCE GRAPH
# ==============================================================

plt.figure(figsize=(13, 7))

plt.plot(

    history_df["Generation"],

    history_df["Training_Loss"],

    label="Training Loss",

    color="#3498db",

    linewidth=2
)

plt.plot(

    history_df["Generation"],

    history_df["Validation_Loss"],

    label="Validation Loss",

    color="#e67e22",

    alpha=0.7
)

plt.plot(

    history_df["Generation"],

    history_df["Best_Validation_Loss"],

    label="Best Immune Memory Loss",

    color="#2ecc71",

    linewidth=2.5
)

plt.axvline(

    best_generation,

    color="red",

    linestyle="--",

    label=f"Best Generation: {best_generation}"
)

plt.title(

    "AIS Clonal Selection - Fitness Convergence",

    fontsize=16,

    fontweight="bold"
)

plt.xlabel("Generation")

plt.ylabel("Scaled Mean Squared Error")

plt.legend()

save_show("ais_convergence_graph.png")

# ==============================================================
# 28. AIS AFFINITY GRAPH
# ==============================================================

plt.figure(figsize=(12, 7))

plt.plot(

    history_df["Generation"],

    history_df["Best_Affinity"],

    color="#2ecc71",

    linewidth=2
)

plt.title(

    "AIS Antibody Affinity Evolution",

    fontsize=16,

    fontweight="bold"
)

plt.xlabel("Generation")

plt.ylabel("Affinity")

save_show("ais_affinity_graph.png")

# ==============================================================
# 29. AIS POPULATION DIVERSITY GRAPH
# ==============================================================

plt.figure(figsize=(12, 7))

plt.plot(

    history_df["Generation"],

    history_df["Population_Diversity"],

    color="#9b59b6",

    linewidth=2
)

plt.title(

    "AIS Antibody Population Diversity",

    fontsize=16,

    fontweight="bold"
)

plt.xlabel("Generation")

plt.ylabel("Mean Parameter Standard Deviation")

save_show("ais_population_diversity_graph.png")

# ==============================================================
# 30. ACTUAL VS PREDICTED SCATTER GRAPH
# ==============================================================

actual = prediction_df[
    "Actual_Required_Workers"
]

predicted = prediction_df[
    "AIS_Predicted_Required_Workers"
]

plt.figure(figsize=(9, 8))

plt.scatter(

    actual,

    predicted,

    alpha=0.5,

    color="#3498db"
)

min_value = min(

    actual.min(),

    predicted.min()
)

max_value = max(

    actual.max(),

    predicted.max()
)

plt.plot(

    [min_value, max_value],

    [min_value, max_value],

    color="red",

    linestyle="--",

    label="Perfect Prediction"
)

plt.title(

    f"AIS Actual vs Predicted Workforce\n"

    f"R² = {test_metrics['R2_Score']:.4f}",

    fontsize=15,

    fontweight="bold"
)

plt.xlabel("Simulated Actual Workforce")

plt.ylabel("AIS Predicted Workforce")

plt.legend()

save_show("ais_actual_vs_predicted.png")

# ==============================================================
# 31. AIS RESIDUAL ERROR GRAPH
# ==============================================================

residuals = actual - predicted

plt.figure(figsize=(12, 7))

plt.scatter(

    predicted,

    residuals,

    alpha=0.5,

    color="#9b59b6"
)

plt.axhline(

    0,

    color="red",

    linestyle="--"
)

plt.title(

    "AIS Workforce Prediction - Residual Analysis",

    fontsize=16,

    fontweight="bold"
)

plt.xlabel("AIS Predicted Workforce")

plt.ylabel("Residual Error")

save_show("ais_residual_graph.png")

# ==============================================================
# 32. AIS ERROR DISTRIBUTION
# ==============================================================

plt.figure(figsize=(12, 7))

sns.histplot(

    residuals,

    bins=40,

    kde=True,

    color="#3498db"
)

plt.axvline(

    0,

    color="red",

    linestyle="--"
)

plt.title(

    "AIS Prediction Error Distribution",

    fontsize=16,

    fontweight="bold"
)

plt.xlabel("Prediction Error")

plt.ylabel("Frequency")

save_show("ais_error_distribution_graph.png")

# ==============================================================
# 33. AIS WORKFORCE ALLOCATION GRAPH
# ==============================================================

allocation_sample = prediction_df.head(25)

x_positions = np.arange(
    len(allocation_sample)
)

width = 0.4

plt.figure(figsize=(16, 7))

plt.bar(

    x_positions - width / 2,

    allocation_sample[
        "available_workers"
    ],

    width,

    color="#3498db",

    label="Available Workforce"
)

plt.bar(

    x_positions + width / 2,

    allocation_sample[
        "AIS_Predicted_Required_Workers"
    ],

    width,

    color="#e67e22",

    label="AIS Required Workforce"
)

plt.xticks(

    x_positions,

    allocation_sample["Scenario_ID"],

    rotation=45
)

plt.title(

    "AIS-Based Municipal Workforce Requirement Analysis",

    fontsize=16,

    fontweight="bold"
)

plt.xlabel("Scenario ID")

plt.ylabel("Number of Workers")

plt.legend()

save_show("ais_workforce_allocation_graph.png")

# ==============================================================
# 34. AIS SHORTAGE AND SURPLUS GRAPH
# ==============================================================

plt.figure(figsize=(16, 7))

plt.bar(

    x_positions - width / 2,

    allocation_sample[
        "AIS_Workforce_Shortage"
    ],

    width,

    color="#e74c3c",

    label="Workforce Shortage"
)

plt.bar(

    x_positions + width / 2,

    allocation_sample[
        "AIS_Workforce_Surplus"
    ],

    width,

    color="#2ecc71",

    label="Workforce Surplus"
)

plt.xticks(

    x_positions,

    allocation_sample["Scenario_ID"],

    rotation=45
)

plt.title(

    "AIS Workforce Shortage and Surplus Analysis",

    fontsize=16,

    fontweight="bold"
)

plt.xlabel("Scenario ID")

plt.ylabel("Number of Workers")

plt.legend()

save_show("ais_workforce_shortage_graph.png")

# ==============================================================
# 35. AIS WORKFORCE STATUS GRAPH
# ==============================================================

status_counts = prediction_df[

    "AIS_Workforce_Status"

].value_counts()

plt.figure(figsize=(9, 8))

status_colors = {

    "Workforce Shortage": "#e74c3c",

    "Workforce Surplus": "#2ecc71",

    "Balanced Workforce": "#3498db"
}

plt.pie(

    status_counts.values,

    labels=status_counts.index,

    autopct="%1.1f%%",

    startangle=90,

    colors=[

        status_colors[name]

        for name in status_counts.index
    ],

    wedgeprops={

        "edgecolor": "white",

        "linewidth": 2
    }
)

plt.title(

    "AIS Predicted Workforce Status Distribution",

    fontsize=16,

    fontweight="bold"
)

save_show("ais_workforce_status_graph.png")

# ==============================================================
# 36. AIS WORKFORCE UTILIZATION GRAPH
# ==============================================================

plt.figure(figsize=(12, 7))

sns.histplot(

    prediction_df[
        "AIS_Workforce_Utilization"
    ],

    bins=25,

    color="#16a085"
)

average_utilization = prediction_df[

    "AIS_Workforce_Utilization"

].mean()

plt.axvline(

    average_utilization,

    color="red",

    linestyle="--",

    label=f"Average: {average_utilization:.2f}%"
)

plt.title(

    "AIS Municipal Workforce Utilization",

    fontsize=16,

    fontweight="bold"
)

plt.xlabel("Workforce Utilization (%)")

plt.ylabel("Number of Scenarios")

plt.legend()

save_show("ais_workforce_utilization_graph.png")

# ==============================================================
# 37. AIS OPERATIONAL COST GRAPH
# ==============================================================

actual_daily_cost = (

    sample[
        "Actual_Required_Workers"
    ]

    * sample[
        "daily_wage"
    ]
)

plt.figure(figsize=(15, 7))

plt.plot(

    positions,

    actual_daily_cost,

    color="#3498db",

    label="Simulated Reference Cost",

    linewidth=2
)

plt.plot(

    positions,

    sample[
        "AIS_Estimated_Daily_Cost"
    ],

    color="#e67e22",

    linestyle="--",

    label="AIS Estimated Cost",

    linewidth=2
)

plt.title(

    "AIS Municipal Workforce - Operational Cost Analysis",

    fontsize=16,

    fontweight="bold"
)

plt.xlabel("Test Scenario")

plt.ylabel("Estimated Daily Cost (INR)")

plt.legend()

save_show("ais_operational_cost_graph.png")

# ==============================================================
# 38. AIS FEATURE IMPORTANCE
# ==============================================================

# Permutation-based feature importance.
# Measures the increase in test MAE after
# shuffling individual input features.

baseline_mae = test_metrics["MAE"]

importance_results = []

for feature_index, feature in enumerate(
    FEATURES
):

    repeated_importance = []

    for repeat in range(5):

        X_permuted = X_test_scaled.copy()

        X_permuted[:, feature_index] = (

            rng.permutation(

                X_permuted[:, feature_index]
            )
        )

        permuted_prediction = (
            predict_workforce(

                X_permuted,

                X_test
            )
        )

        permuted_mae = mean_absolute_error(

            y_test,

            permuted_prediction
        )

        repeated_importance.append(

            permuted_mae - baseline_mae
        )

    importance_results.append({

        "Feature": feature,

        "Importance": float(
            np.mean(repeated_importance)
        ),

        "Standard_Deviation": float(
            np.std(repeated_importance)
        )
    })

importance_df = pd.DataFrame(
    importance_results
)

importance_df = importance_df.sort_values(

    "Importance",

    ascending=True
)

importance_df.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "ais_feature_importance.csv"
    ),

    index=False
)

plt.figure(figsize=(12, 8))

plt.barh(

    importance_df["Feature"],

    importance_df["Importance"],

    xerr=importance_df[
        "Standard_Deviation"
    ],

    color="#2ecc71",

    edgecolor="black"
)

plt.title(

    "AIS Permutation Feature Importance",

    fontsize=16,

    fontweight="bold"
)

plt.xlabel("Increase in MAE After Permutation")

plt.axvline(

    0,

    color="black"
)

save_show("ais_feature_importance_graph.png")

# ==============================================================
# 39. AIS WORKFORCE SUMMARY
# ==============================================================

summary = {

    "Algorithm": "Artificial Immune System",

    "Optimization_Method": "Clonal Selection",

    "Total_Test_Scenarios": int(
        len(prediction_df)
    ),

    "Best_Generation": int(
        best_generation
    ),

    "Best_Validation_Loss": float(
        best_validation_loss
    ),

    "R2_Score": test_metrics[
        "R2_Score"
    ],

    "R2_Percentage": test_metrics[
        "R2_Percentage"
    ],

    "MAE": test_metrics[
        "MAE"
    ],

    "RMSE": test_metrics[
        "RMSE"
    ],

    "MAPE_Percentage": test_metrics[
        "MAPE_Percentage"
    ],

    "Average_Available_Workforce": float(

        prediction_df[
            "available_workers"
        ].mean()
    ),

    "Average_Predicted_Workforce": float(

        prediction_df[
            "AIS_Predicted_Required_Workers"
        ].mean()
    ),

    "Average_Workforce_Shortage": float(

        prediction_df[
            "AIS_Workforce_Shortage"
        ].mean()
    ),

    "Average_Workforce_Surplus": float(

        prediction_df[
            "AIS_Workforce_Surplus"
        ].mean()
    ),

    "Average_Workforce_Utilization": float(

        prediction_df[
            "AIS_Workforce_Utilization"
        ].mean()
    ),

    "Average_Daily_Cost": float(

        prediction_df[
            "AIS_Estimated_Daily_Cost"
        ].mean()
    ),

    "Shortage_Scenarios": int(

        (

            prediction_df[
                "AIS_Workforce_Status"
            ]

            == "Workforce Shortage"

        ).sum()
    ),

    "Surplus_Scenarios": int(

        (

            prediction_df[
                "AIS_Workforce_Status"
            ]

            == "Workforce Surplus"

        ).sum()
    ),

    "Balanced_Scenarios": int(

        (

            prediction_df[
                "AIS_Workforce_Status"
            ]

            == "Balanced Workforce"

        ).sum()
    ),

    "Execution_Time_Seconds": float(
        time.time() - START_TIME
    ),

    "Evaluation_Type": "Synthetic Simulation",

    "Real_World_Validated": False
}

pd.DataFrame([summary]).to_csv(

    os.path.join(
        OUTPUT_DIR,
        "ais_workforce_summary.csv"
    ),

    index=False
)

# ==============================================================
# 40. SAVE AIS YAML CONFIGURATION
# ==============================================================

configuration = {

    "project": {

        "name": (
            "AIS-Based Municipal Waste Management "
            "and Conservancy Workforce Optimization"
        ),

        "version": "1.0",

        "algorithm": "Artificial Immune System",

        "method": "Clonal Selection Algorithm",

        "random_seed": SEED,

        "data_type": "synthetic_simulation"
    },

    "dataset": {

        "original_dataset": ORIGINAL_DATASET,

        "synthetic_dataset": SYNTHETIC_DATASET,

        "total_records": int(len(df)),

        "training_records": int(len(X_train)),

        "validation_records": int(len(X_val)),

        "testing_records": int(len(X_test)),

        "features": FEATURES,

        "target": TARGET
    },

    "ais_parameters": {

        "population_size": POPULATION_SIZE,

        "generations": GENERATIONS,

        "selected_antibodies": SELECTED_ANTIBODIES,

        "base_clones_per_antibody": CLONES_PER_ANTIBODY,

        "receptor_editing_rate": RECEPTOR_EDITING_RATE,

        "initial_mutation_rate": INITIAL_MUTATION_RATE,

        "minimum_mutation_rate": MIN_MUTATION_RATE,

        "weight_limit": WEIGHT_LIMIT,

        "regularization": REGULARIZATION,

        "fitness_function": (
            "Training MSE with weight regularization"
        ),

        "affinity_function": "1 / (1 + fitness)",

        "memory_selection": "Minimum validation MSE"
    },

    "neural_network": {

        "input_size": INPUT_SIZE,

        "hidden_layer_1": HIDDEN_LAYER_1,

        "hidden_layer_2": HIDDEN_LAYER_2,

        "output_size": 1,

        "activation": "ReLU",

        "total_parameters": TOTAL_PARAMETERS,

        "gradient_training": False
    },

    "constraints": {

        "minimum_workers_per_site": 2,

        "integer_workforce_predictions": True
    },

    "files": {

        "h5_model": "ais_model.h5",

        "pkl_model": "ais_model.pkl",

        "feature_scaler": "ais_scaler.pkl",

        "target_scaler": "ais_target_scaler.pkl",

        "best_antibody": "ais_best_antibody.npy",

        "results": "ais_result.csv",

        "predictions": "ais_prediction.csv"
    },

    "limitations": [

        "Original dataset contains aggregate statistics.",

        "Training and testing use synthetic scenarios.",

        "AIS optimizes prediction model parameters.",

        "Site-level workforce allocation is not validated.",

        "Real-world municipal performance is not established."
    ]
}

with open(

    os.path.join(
        OUTPUT_DIR,
        "ais_config.yaml"
    ),

    "w",

    encoding="utf-8"

) as file:

    yaml.safe_dump(

        configuration,

        file,

        sort_keys=False,

        allow_unicode=True
    )

# ==============================================================
# 41. SAVE AIS JSON METRICS
# ==============================================================

metrics_report = {

    "project": (

        "AIS-Based Municipal Waste Management "

        "and Conservancy Workforce Optimization"
    ),

    "algorithm": "Artificial Immune System",

    "method": "Clonal Selection Algorithm",

    "evaluation_type": "Synthetic Simulation",

    "best_generation": best_generation,

    "best_validation_loss": float(
        best_validation_loss
    ),

    "training_metrics": train_metrics,

    "validation_metrics": validation_metrics,

    "test_metrics": test_metrics,

    "workforce_summary": summary,

    "real_world_validated": False
}

with open(

    os.path.join(
        OUTPUT_DIR,
        "ais_metrics.json"
    ),

    "w",

    encoding="utf-8"

) as file:

    json.dump(

        metrics_report,

        file,

        indent=4,

        allow_nan=False
    )

# Save neural network architecture.

with open(

    os.path.join(
        OUTPUT_DIR,
        "ais_model_architecture.json"
    ),

    "w",

    encoding="utf-8"

) as file:

    file.write(
        ais_model.to_json()
    )

# ==============================================================
# 42. PREDICT A NEW MUNICIPAL SCENARIO
# ==============================================================

new_scenario = pd.DataFrame([{

    "disposal_sites": 105,

    "other_workers": 412,

    "daily_waste_tonnes": 650,

    "population_density": 12000,

    "working_hours": 8,

    "worker_productivity": 1.0,

    "equipment_efficiency": 0.85,

    "absenteeism_rate": 0.08,

    "transport_distance_km": 10,

    "seasonal_demand_factor": 1.1
}])

new_scaled = feature_scaler.transform(

    new_scenario[FEATURES]

).astype(np.float32)

new_prediction = int(

    predict_workforce(

        new_scaled,

        new_scenario

    )[0]
)

available_workers = 847

shortage = max(

    new_prediction - available_workers,

    0
)

surplus = max(

    available_workers - new_prediction,

    0
)

new_scenario["Available_Workers"] = (
    available_workers
)

new_scenario["AIS_Predicted_Workers"] = (
    new_prediction
)

new_scenario["AIS_Workforce_Shortage"] = (
    shortage
)

new_scenario["AIS_Workforce_Surplus"] = (
    surplus
)

new_scenario["Data_Type"] = (
    "Synthetic Simulation"
)

new_scenario.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "ais_new_scenario_prediction.csv"
    ),

    index=False
)

print("\n" + "=" * 75)
print("AIS NEW MUNICIPAL SCENARIO")
print("=" * 75)

print("Available Workers:", available_workers)

print("Predicted Workers:", new_prediction)

print("Workforce Shortage:", shortage)

print("Workforce Surplus:", surplus)

# ==============================================================
# 43. DISPLAY FINAL SUMMARY
# ==============================================================

print("\n" + "=" * 75)
print("FINAL AIS PERFORMANCE SUMMARY")
print("=" * 75)

for key, value in summary.items():

    if isinstance(value, float):

        print(
            f"{key:<40}: {value:.4f}"
        )

    else:

        print(
            f"{key:<40}: {value}"
        )

# ==============================================================
# 44. DISPLAY GENERATED AIS FILES
# ==============================================================

print("\n" + "=" * 75)
print("GENERATED AIS OUTPUT FILES")
print("=" * 75)

file_count = 0

for filename in sorted(

    os.listdir(OUTPUT_DIR)
):

    if filename.startswith("ais_"):

        filepath = os.path.join(

            OUTPUT_DIR,

            filename
        )

        if os.path.isfile(filepath):

            size_kb = (

                os.path.getsize(filepath)

                / 1024
            )

            print(

                f"{filename:<45} "

                f"{size_kb:>10.2f} KB"
            )

            file_count += 1

print("\nTotal AIS Files:", file_count)

print("\nOutput Directory:")

print(OUTPUT_DIR)

print("\nTotal Execution Time:")

print(

    f"{time.time() - START_TIME:.2f} seconds"
)

print("\nAIS PROJECT COMPLETED SUCCESSFULLY.")

print(

    "\nIMPORTANT: All reported results are based "

    "on synthetic operational scenarios and "

    "are not validated real-world municipal predictions."
)
```

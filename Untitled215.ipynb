```python
import os
import json
import yaml
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from tensorflow.keras.models import load_model

# ==============================================================
# 1. CONFIGURATION
# ==============================================================

OUTPUT_DIR = r"C:\Users\sagni\Downloads\AI-Based Municipal Waste Management and Conservancy Workforce"

DATASET_PATH = os.path.join(
    OUTPUT_DIR,
    "synthetic_municipal_dataset.csv"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

sns.set_theme(style="whitegrid")

plt.rcParams["figure.dpi"] = 120
plt.rcParams["savefig.dpi"] = 300
plt.rcParams["font.size"] = 11

print("=" * 75)
print("MUNICIPAL WASTE MANAGEMENT - PERFORMANCE EVALUATION")
print("=" * 75)

# ==============================================================
# 2. LOAD CONFIGURATION AND DATASET
# ==============================================================

with open(
    os.path.join(OUTPUT_DIR, "config.yaml"),
    "r",
    encoding="utf-8"
) as file:

    config = yaml.safe_load(file)

with open(
    os.path.join(OUTPUT_DIR, "metrics.json"),
    "r",
    encoding="utf-8"
) as file:

    previous_metrics = json.load(file)

df = pd.read_csv(DATASET_PATH)

FEATURES = config["dataset"]["features"]
TARGET = config["dataset"]["target"]
SEED = config["project"]["random_seed"]

print("\nDataset Shape:", df.shape)
print("\nFeatures:", FEATURES)
print("\nTarget:", TARGET)

print("\nDataset Preview:")
print(df.head())

# ==============================================================
# 3. LOAD TRAINED MODELS
# ==============================================================

scaler = joblib.load(
    os.path.join(OUTPUT_DIR, "scaler.pkl")
)

target_scaler = joblib.load(
    os.path.join(OUTPUT_DIR, "target_scaler.pkl")
)

models = {

    "Linear Regression": joblib.load(
        os.path.join(
            OUTPUT_DIR,
            "linear_regression_model.pkl"
        )
    ),

    "Random Forest": joblib.load(
        os.path.join(
            OUTPUT_DIR,
            "random_forest_model.pkl"
        )
    ),

    "Gradient Boosting": joblib.load(
        os.path.join(
            OUTPUT_DIR,
            "gradient_boosting_model.pkl"
        )
    ),

    "Extra Trees": joblib.load(
        os.path.join(
            OUTPUT_DIR,
            "extra_trees_model.pkl"
        )
    )
}

ann_model = load_model(
    os.path.join(OUTPUT_DIR, "model.h5"),
    compile=False
)

print("\nAll trained models loaded successfully.")

# ==============================================================
# 4. RECREATE ORIGINAL TRAIN-VALIDATION-TEST SPLIT
# ==============================================================

X = df[FEATURES]
y = df[TARGET]

X_train_full, X_test, y_train_full, y_test = (
    train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=SEED
    )
)

X_train, X_val, y_train, y_val = (
    train_test_split(
        X_train_full,
        y_train_full,
        test_size=0.20,
        random_state=SEED
    )
)

X_train_scaled = scaler.transform(X_train)
X_val_scaled = scaler.transform(X_val)
X_test_scaled = scaler.transform(X_test)

print("\nTraining Samples:", len(X_train))
print("Validation Samples:", len(X_val))
print("Testing Samples:", len(X_test))

# ==============================================================
# 5. PERFORMANCE EVALUATION FUNCTION
# ==============================================================

def evaluate_model(actual, predicted):

    actual = np.asarray(actual).flatten()

    predicted = np.asarray(predicted).flatten()

    mae = mean_absolute_error(
        actual, predicted
    )

    mse = mean_squared_error(
        actual, predicted
    )

    rmse = np.sqrt(mse)

    r2 = r2_score(
        actual, predicted
    )

    mape = np.mean(
        np.abs(
            (actual - predicted)
            / np.maximum(np.abs(actual), 1)
        )
    ) * 100

    return {

        "MAE": float(mae),

        "MSE": float(mse),

        "RMSE": float(rmse),

        "R2_Score": float(r2),

        "R2_Percentage": float(r2 * 100),

        "MAPE_Percentage": float(mape)
    }

# ==============================================================
# 6. ENFORCE WORKFORCE CONSTRAINTS
# ==============================================================

def enforce_constraints(predictions, features):

    minimum_workers = (
        2 * features["disposal_sites"].to_numpy()
    )

    predictions = np.asarray(
        predictions
    ).flatten()

    predictions = np.maximum(
        np.ceil(predictions),
        minimum_workers
    )

    return predictions.astype(int)

# ==============================================================
# 7. GENERATE MODEL PREDICTIONS
# ==============================================================

predictions = {}
validation_predictions = {}
results = {}
validation_results = {}

for name, model in models.items():

    test_pred = model.predict(
        X_test_scaled
    )

    val_pred = model.predict(
        X_val_scaled
    )

    predictions[name] = enforce_constraints(
        test_pred,
        X_test
    )

    validation_predictions[name] = enforce_constraints(
        val_pred,
        X_val
    )

    results[name] = evaluate_model(
        y_test,
        predictions[name]
    )

    validation_results[name] = evaluate_model(
        y_val,
        validation_predictions[name]
    )

# ANN predictions

ann_test_pred = target_scaler.inverse_transform(
    ann_model.predict(
        X_test_scaled,
        verbose=0
    )
).flatten()

ann_val_pred = target_scaler.inverse_transform(
    ann_model.predict(
        X_val_scaled,
        verbose=0
    )
).flatten()

predictions["ANN"] = enforce_constraints(
    ann_test_pred,
    X_test
)

validation_predictions["ANN"] = enforce_constraints(
    ann_val_pred,
    X_val
)

results["ANN"] = evaluate_model(
    y_test,
    predictions["ANN"]
)

validation_results["ANN"] = evaluate_model(
    y_val,
    validation_predictions["ANN"]
)

# Select best model using validation performance only

best_model_name = max(
    validation_results,
    key=lambda name:
        validation_results[name]["R2_Score"]
)

best_predictions = predictions[best_model_name]

print("\nBest Model:", best_model_name)

# ==============================================================
# 8. CREATE AND SAVE RESULT CSV
# ==============================================================

result_df = pd.DataFrame(results).T

result_df.index.name = "Model"

result_df = result_df.reset_index()

result_df["Validation_R2"] = (
    result_df["Model"].map(
        lambda name:
            validation_results[name]["R2_Score"]
    )
)

result_df["Selected_Model"] = (
    result_df["Model"] == best_model_name
)

result_df["Evaluation_Type"] = (
    "Synthetic Simulation"
)

result_df = result_df.sort_values(
    "R2_Score",
    ascending=False
).reset_index(drop=True)

result_df.to_csv(
    os.path.join(OUTPUT_DIR, "result.csv"),
    index=False
)

print("\n" + "=" * 75)
print("MODEL PERFORMANCE RESULTS")
print("=" * 75)

print(result_df.round(4).to_string(index=False))

# ==============================================================
# 9. FUNCTION TO SAVE AND SHOW GRAPHS
# ==============================================================

def save_show(filename):

    plt.tight_layout()

    plt.savefig(
        os.path.join(OUTPUT_DIR, filename),
        dpi=300,
        bbox_inches="tight",
        facecolor="white"
    )

    plt.show()
    plt.close()

    print("Saved:", filename)

# ==============================================================
# 10. ACCURACY GRAPH - REGRESSION R2 PERCENTAGE
# ==============================================================

plt.figure(figsize=(12, 7))

colors = [
    "#3498db",
    "#2ecc71",
    "#e67e22",
    "#9b59b6",
    "#e74c3c"
]

bars = plt.bar(
    result_df["Model"],
    result_df["R2_Percentage"],
    color=colors,
    edgecolor="black",
    alpha=0.85
)

for bar, value in zip(
    bars,
    result_df["R2_Percentage"]
):

    plt.annotate(
        f"{value:.2f}%",
        (
            bar.get_x() + bar.get_width() / 2,
            value
        ),
        ha="center",
        va="bottom" if value >= 0 else "top",
        xytext=(0, 5 if value >= 0 else -5),
        textcoords="offset points",
        fontweight="bold"
    )

plt.title(
    "Municipal Workforce Prediction - Model Performance",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Machine Learning Models")
plt.ylabel("R² Score (%)")

plt.axhline(
    0,
    color="black",
    linewidth=1
)

plt.ylim(
    min(
        0,
        result_df["R2_Percentage"].min() - 10
    ),
    max(
        105,
        result_df["R2_Percentage"].max() + 10
    )
)

plt.xticks(rotation=15)
plt.grid(axis="y", alpha=0.3)

save_show("accuracy_graph.png")

# ==============================================================
# 11. CORRELATION HEATMAP
# ==============================================================

correlation_columns = FEATURES + [TARGET]

correlation_matrix = (
    df[correlation_columns].corr()
)

plt.figure(figsize=(15, 11))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1,
    linewidths=0.5,
    square=True,
    cbar_kws={
        "label": "Correlation Coefficient"
    }
)

plt.title(
    "Municipal Waste Management - Feature Correlation Heatmap",
    fontsize=16,
    fontweight="bold"
)

plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)

save_show("heatmap.png")

correlation_matrix.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "correlation_matrix.csv"
    )
)

# ==============================================================
# 12. MODEL COMPARISON GRAPH
# ==============================================================

fig, axes = plt.subplots(
    2, 2,
    figsize=(16, 11)
)

metrics_to_plot = [
    ("MAE", "Mean Absolute Error", "#3498db"),
    ("RMSE", "Root Mean Squared Error", "#e67e22"),
    ("R2_Score", "R² Score", "#2ecc71"),
    ("MAPE_Percentage", "MAPE (%)", "#9b59b6")
]

for ax, (metric, title, color) in zip(
    axes.flatten(),
    metrics_to_plot
):

    bars = ax.bar(
        result_df["Model"],
        result_df[metric],
        color=color,
        edgecolor="black",
        alpha=0.85
    )

    ax.set_title(
        title,
        fontsize=14,
        fontweight="bold"
    )

    ax.set_ylabel(metric)

    ax.tick_params(
        axis="x",
        rotation=20
    )

    ax.margins(y=0.15)

    for bar, value in zip(
        bars,
        result_df[metric]
    ):

        ax.annotate(
            f"{value:.3f}",
            (
                bar.get_x() + bar.get_width() / 2,
                value
            ),
            ha="center",
            va="bottom" if value >= 0 else "top",
            fontsize=9
        )

    ax.grid(axis="y", alpha=0.3)

fig.suptitle(
    "Municipal Workforce Prediction - Model Comparison",
    fontsize=18,
    fontweight="bold"
)

save_show("comparison_graph.png")

# ==============================================================
# 13. RESULT GRAPH - TRAINING VALIDATION VS TEST
# ==============================================================

positions = np.arange(
    len(result_df)
)

width = 0.35

plt.figure(figsize=(13, 7))

plt.bar(
    positions - width / 2,
    result_df["Validation_R2"] * 100,
    width,
    label="Validation R²",
    color="#3498db",
    edgecolor="black"
)

plt.bar(
    positions + width / 2,
    result_df["R2_Percentage"],
    width,
    label="Testing R²",
    color="#2ecc71",
    edgecolor="black"
)

plt.xticks(
    positions,
    result_df["Model"],
    rotation=15
)

plt.xlabel("Machine Learning Models")
plt.ylabel("R² Score (%)")

plt.title(
    "Validation and Testing Performance Comparison",
    fontsize=16,
    fontweight="bold"
)

plt.legend()
plt.grid(axis="y", alpha=0.3)

save_show("result_graph.png")

# ==============================================================
# 14. CREATE PREDICTION DATAFRAME
# ==============================================================

prediction_df = df.loc[
    X_test.index
].copy()

prediction_df.insert(
    0,
    "Scenario_ID",
    X_test.index.to_numpy() + 1
)

prediction_df["Actual_Required_Workers"] = (
    y_test.to_numpy()
)

for name, values in predictions.items():

    column_name = (
        name.replace(" ", "_") + "_Prediction"
    )

    prediction_df[column_name] = values

prediction_df["Predicted_Required_Workers"] = (
    best_predictions
)

prediction_df["Selected_Model"] = (
    best_model_name
)

prediction_df["Absolute_Error"] = np.abs(

    prediction_df["Actual_Required_Workers"]

    - prediction_df["Predicted_Required_Workers"]
)

prediction_df["Percentage_Error"] = (

    prediction_df["Absolute_Error"]

    / np.maximum(
        prediction_df["Actual_Required_Workers"],
        1
    )

) * 100

# ==============================================================
# 15. WORKFORCE SHORTAGE AND SURPLUS
# ==============================================================

prediction_df["Predicted_Workforce_Shortage"] = (
    np.maximum(

        prediction_df["Predicted_Required_Workers"]

        - prediction_df["available_workers"],

        0
    )
)

prediction_df["Predicted_Workforce_Surplus"] = (
    np.maximum(

        prediction_df["available_workers"]

        - prediction_df["Predicted_Required_Workers"],

        0
    )
)

prediction_df["Predicted_Workforce_Utilization"] = (

    np.minimum(

        prediction_df["Predicted_Required_Workers"]

        / np.maximum(
            prediction_df["available_workers"],
            1
        ),

        1.0
    )

    * 100
)

prediction_df["Predicted_Daily_Cost"] = (

    prediction_df["Predicted_Required_Workers"]

    * prediction_df["daily_wage"]
)

prediction_df["Additional_Workforce_Cost"] = (

    prediction_df["Predicted_Workforce_Shortage"]

    * prediction_df["daily_wage"]
)

prediction_df["Workforce_Status"] = np.select(

    [
        prediction_df[
            "Predicted_Workforce_Shortage"
        ] > 0,

        prediction_df[
            "Predicted_Workforce_Surplus"
        ] > 0
    ],

    [
        "Workforce Shortage",
        "Workforce Surplus"
    ],

    default="Balanced Workforce"
)

prediction_df["Recommendation"] = np.select(

    [
        prediction_df[
            "Predicted_Workforce_Shortage"
        ] > 0,

        prediction_df[
            "Predicted_Workforce_Surplus"
        ] > 0
    ],

    [
        "Allocate Additional Workers",
        "Consider Workforce Redistribution"
    ],

    default="Maintain Current Workforce"
)

prediction_df["Data_Type"] = (
    "Synthetic Simulation"
)

prediction_df.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "prediction.csv"
    ),
    index=False
)

print("\nPrediction Results:")

print(
    prediction_df[[
        "Scenario_ID",
        "available_workers",
        "Actual_Required_Workers",
        "Predicted_Required_Workers",
        "Absolute_Error",
        "Workforce_Status",
        "Recommendation"
    ]].head(10).to_string(index=False)
)

# ==============================================================
# 16. PREDICTION GRAPH
# ==============================================================

N_DISPLAY = min(
    100,
    len(prediction_df)
)

sample = prediction_df.head(
    N_DISPLAY
)

plt.figure(figsize=(16, 7))

plt.plot(
    np.arange(1, N_DISPLAY + 1),
    sample["Actual_Required_Workers"],
    label="Simulated Actual Workforce",
    color="#3498db",
    linewidth=2,
    marker="o",
    markersize=3
)

plt.plot(
    np.arange(1, N_DISPLAY + 1),
    sample["Predicted_Required_Workers"],
    label="Predicted Workforce",
    color="#e74c3c",
    linewidth=2,
    linestyle="--",
    marker="x",
    markersize=3
)

plt.title(
    f"Actual vs Predicted Workforce Requirements - {best_model_name}",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Test Scenario")
plt.ylabel("Required Workers")
plt.legend()
plt.grid(alpha=0.3)

save_show("prediction_graph.png")

# ==============================================================
# 17. ACTUAL VS PREDICTED SCATTER GRAPH
# ==============================================================

plt.figure(figsize=(9, 8))

plt.scatter(

    prediction_df["Actual_Required_Workers"],

    prediction_df["Predicted_Required_Workers"],

    color="#3498db",

    alpha=0.5,

    edgecolors="none"
)

min_value = min(
    prediction_df["Actual_Required_Workers"].min(),
    prediction_df["Predicted_Required_Workers"].min()
)

max_value = max(
    prediction_df["Actual_Required_Workers"].max(),
    prediction_df["Predicted_Required_Workers"].max()
)

plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    color="red",
    linestyle="--",
    linewidth=2,
    label="Perfect Prediction"
)

plt.xlabel("Simulated Actual Workers")
plt.ylabel("Predicted Workers")

plt.title(
    f"Actual vs Predicted Workforce\n"
    f"R² = {results[best_model_name]['R2_Score']:.4f}",
    fontsize=15,
    fontweight="bold"
)

plt.legend()
plt.grid(alpha=0.3)

save_show("actual_vs_predicted.png")

# ==============================================================
# 18. WORKFORCE ALLOCATION GRAPH
# ==============================================================

allocation_sample = prediction_df.head(25)

positions = np.arange(
    len(allocation_sample)
)

width = 0.4

plt.figure(figsize=(16, 7))

plt.bar(
    positions - width / 2,
    allocation_sample["available_workers"],
    width,
    color="#3498db",
    label="Available Workforce"
)

plt.bar(
    positions + width / 2,
    allocation_sample["Predicted_Required_Workers"],
    width,
    color="#e67e22",
    label="Required Workforce"
)

plt.xticks(
    positions,
    allocation_sample["Scenario_ID"],
    rotation=45
)

plt.xlabel("Scenario ID")
plt.ylabel("Number of Workers")

plt.title(
    "Available vs Required Municipal Workforce",
    fontsize=16,
    fontweight="bold"
)

plt.legend()
plt.grid(axis="y", alpha=0.3)

save_show("workforce_allocation_graph.png")

# ==============================================================
# 19. WORKFORCE SHORTAGE AND SURPLUS GRAPH
# ==============================================================

plt.figure(figsize=(15, 7))

plt.bar(

    positions - width / 2,

    allocation_sample[
        "Predicted_Workforce_Shortage"
    ],

    width,

    color="#e74c3c",

    label="Workforce Shortage"
)

plt.bar(

    positions + width / 2,

    allocation_sample[
        "Predicted_Workforce_Surplus"
    ],

    width,

    color="#2ecc71",

    label="Workforce Surplus"
)

plt.xticks(
    positions,
    allocation_sample["Scenario_ID"],
    rotation=45
)

plt.xlabel("Scenario ID")
plt.ylabel("Number of Workers")

plt.title(
    "Municipal Workforce Shortage and Surplus Analysis",
    fontsize=16,
    fontweight="bold"
)

plt.legend()

save_show("workforce_shortage_graph.png")

# ==============================================================
# 20. WORKFORCE STATUS DISTRIBUTION
# ==============================================================

status_counts = (
    prediction_df["Workforce_Status"]
    .value_counts()
)

status_colors = {

    "Workforce Shortage": "#e74c3c",

    "Workforce Surplus": "#2ecc71",

    "Balanced Workforce": "#3498db"
}

plt.figure(figsize=(9, 8))

plt.pie(

    status_counts.values,

    labels=status_counts.index,

    autopct="%1.1f%%",

    startangle=90,

    colors=[
        status_colors[name]
        for name in status_counts.index
    ],

    wedgeprops={
        "edgecolor": "white",
        "linewidth": 2
    }
)

plt.title(
    "Predicted Workforce Status Distribution",
    fontsize=16,
    fontweight="bold"
)

save_show("workforce_status_graph.png")

# ==============================================================
# 21. WORKFORCE UTILIZATION GRAPH
# ==============================================================

plt.figure(figsize=(12, 7))

sns.histplot(

    prediction_df[
        "Predicted_Workforce_Utilization"
    ],

    bins=25,

    color="#16a085",

    kde=False
)

average_utilization = prediction_df[
    "Predicted_Workforce_Utilization"
].mean()

plt.axvline(

    average_utilization,

    color="red",

    linestyle="--",

    linewidth=2,

    label=f"Average: {average_utilization:.2f}%"
)

plt.title(
    "Municipal Workforce Utilization Distribution",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Workforce Utilization (%)")
plt.ylabel("Number of Scenarios")
plt.legend()

save_show("workforce_utilization_graph.png")

# ==============================================================
# 22. OPERATIONAL COST GRAPH
# ==============================================================

plt.figure(figsize=(15, 7))

plt.plot(

    np.arange(1, N_DISPLAY + 1),

    sample["estimated_daily_cost"],

    color="#3498db",

    label="Simulated Actual Daily Cost",

    linewidth=2
)

plt.plot(

    np.arange(1, N_DISPLAY + 1),

    sample["Predicted_Daily_Cost"],

    color="#e67e22",

    linestyle="--",

    label="Predicted Daily Cost",

    linewidth=2
)

plt.title(
    "Municipal Waste Management - Daily Workforce Cost",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Test Scenario")
plt.ylabel("Estimated Daily Cost (INR)")
plt.legend()

save_show("operational_cost_graph.png")

# ==============================================================
# 23. RESIDUAL ERROR GRAPH
# ==============================================================

residuals = (

    prediction_df["Actual_Required_Workers"]

    - prediction_df["Predicted_Required_Workers"]
)

plt.figure(figsize=(11, 7))

plt.scatter(

    prediction_df["Predicted_Required_Workers"],

    residuals,

    alpha=0.5,

    color="#9b59b6"
)

plt.axhline(
    0,
    color="red",
    linestyle="--",
    linewidth=2
)

plt.title(
    "Workforce Prediction - Residual Error Analysis",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Predicted Workforce")
plt.ylabel("Residual Error")

save_show("residual_graph.png")

# ==============================================================
# 24. ERROR DISTRIBUTION GRAPH
# ==============================================================

plt.figure(figsize=(11, 7))

sns.histplot(
    residuals,
    bins=40,
    kde=True,
    color="#3498db"
)

plt.axvline(
    0,
    color="red",
    linestyle="--",
    linewidth=2
)

plt.title(
    "Workforce Prediction Error Distribution",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Prediction Error")
plt.ylabel("Frequency")

save_show("error_distribution_graph.png")

# ==============================================================
# 25. FEATURE IMPORTANCE GRAPH
# ==============================================================

rf_model = models["Random Forest"]

importance_df = pd.DataFrame({

    "Feature": FEATURES,

    "Importance": rf_model.feature_importances_
})

importance_df = importance_df.sort_values(
    "Importance",
    ascending=True
)

plt.figure(figsize=(12, 8))

plt.barh(

    importance_df["Feature"],

    importance_df["Importance"],

    color="#2ecc71",

    edgecolor="black"
)

plt.title(
    "Random Forest - Workforce Prediction Feature Importance",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Feature Importance")
plt.ylabel("Operational Features")

save_show("feature_importance_graph.png")

importance_df.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "feature_importance.csv"
    ),
    index=False
)

# ==============================================================
# 26. ANN TRAINING HISTORY GRAPH
# ==============================================================

history_path = os.path.join(
    OUTPUT_DIR,
    "training_history.csv"
)

if os.path.exists(history_path):

    history_df = pd.read_csv(
        history_path
    )

    epochs = np.arange(
        1,
        len(history_df) + 1
    )

    plt.figure(figsize=(12, 7))

    plt.plot(
        epochs,
        history_df["loss"],
        label="Training Loss",
        color="#3498db",
        linewidth=2
    )

    plt.plot(
        epochs,
        history_df["val_loss"],
        label="Validation Loss",
        color="#e74c3c",
        linewidth=2
    )

    plt.title(
        "Artificial Neural Network - Training and Validation Loss",
        fontsize=16,
        fontweight="bold"
    )

    plt.xlabel("Epoch")
    plt.ylabel("Mean Squared Error (Scaled Target)")
    plt.legend()
    plt.grid(alpha=0.3)

    save_show("training_loss_graph.png")

# ==============================================================
# 27. MUNICIPAL WORKFORCE SUMMARY CSV
# ==============================================================

summary = {

    "Total_Test_Scenarios": int(
        len(prediction_df)
    ),

    "Best_Model": best_model_name,

    "R2_Score": float(
        results[best_model_name]["R2_Score"]
    ),

    "R2_Percentage": float(
        results[best_model_name]["R2_Percentage"]
    ),

    "MAE": float(
        results[best_model_name]["MAE"]
    ),

    "RMSE": float(
        results[best_model_name]["RMSE"]
    ),

    "MAPE_Percentage": float(
        results[best_model_name]["MAPE_Percentage"]
    ),

    "Average_Available_Workforce": float(
        prediction_df["available_workers"].mean()
    ),

    "Average_Required_Workforce": float(
        prediction_df[
            "Predicted_Required_Workers"
        ].mean()
    ),

    "Average_Workforce_Shortage": float(
        prediction_df[
            "Predicted_Workforce_Shortage"
        ].mean()
    ),

    "Average_Workforce_Surplus": float(
        prediction_df[
            "Predicted_Workforce_Surplus"
        ].mean()
    ),

    "Average_Workforce_Utilization": float(
        prediction_df[
            "Predicted_Workforce_Utilization"
        ].mean()
    ),

    "Average_Daily_Workforce_Cost": float(
        prediction_df[
            "Predicted_Daily_Cost"
        ].mean()
    ),

    "Shortage_Scenarios": int(
        (
            prediction_df["Workforce_Status"]
            == "Workforce Shortage"
        ).sum()
    ),

    "Surplus_Scenarios": int(
        (
            prediction_df["Workforce_Status"]
            == "Workforce Surplus"
        ).sum()
    ),

    "Balanced_Scenarios": int(
        (
            prediction_df["Workforce_Status"]
            == "Balanced Workforce"
        ).sum()
    ),

    "Evaluation_Type": "Synthetic Simulation",

    "Real_World_Validated": False
}

summary_df = pd.DataFrame(
    [summary]
)

summary_df.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "workforce_summary.csv"
    ),
    index=False
)

# ==============================================================
# 28. SAVE EVALUATION JSON
# ==============================================================

evaluation_report = {

    "project": (
        "AI-Based Municipal Waste Management "
        "and Conservancy Workforce"
    ),

    "evaluation_type": "Synthetic Simulation",

    "best_model": best_model_name,

    "test_results": results,

    "validation_results": validation_results,

    "workforce_summary": summary,

    "real_world_validated": False
}

with open(

    os.path.join(
        OUTPUT_DIR,
        "evaluation_results.json"
    ),

    "w",

    encoding="utf-8"

) as file:

    json.dump(
        evaluation_report,
        file,
        indent=4,
        allow_nan=False
    )

# ==============================================================
# 29. DISPLAY FINAL SUMMARY
# ==============================================================

print("\n" + "=" * 75)
print("FINAL PROJECT SUMMARY")
print("=" * 75)

for key, value in summary.items():

    if isinstance(value, float):

        print(f"{key:<40}: {value:.4f}")

    else:

        print(f"{key:<40}: {value}")

# ==============================================================
# 30. DISPLAY GENERATED FILES
# ==============================================================

print("\n" + "=" * 75)
print("GENERATED OUTPUT FILES")
print("=" * 75)

for filename in sorted(
    os.listdir(OUTPUT_DIR)
):

    if filename.endswith(
        (".png", ".csv", ".json", ".yaml", ".pkl", ".h5")
    ):

        filepath = os.path.join(
            OUTPUT_DIR,
            filename
        )

        size = os.path.getsize(
            filepath
        ) / 1024

        print(
            f"{filename:<45} "
            f"{size:>10.2f} KB"
        )

print("\nOutput Directory:")
print(OUTPUT_DIR)

print("\nAll results, predictions and graphs saved successfully.")

print(
    "\nNOTE: All evaluation results represent "
    "synthetic operational scenarios, not "
    "validated real-world municipal performance."
)
```
